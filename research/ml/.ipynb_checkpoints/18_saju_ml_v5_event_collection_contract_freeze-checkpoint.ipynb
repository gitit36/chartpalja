# 18 — V5 DEV Event Collection Contract & Batch Freeze

This notebook freezes the event-research protocol and deterministic 8-batch worklist **before any event research**.

It does not access event outcomes, chronology, astrology, Control, or CONFIRM.


```python

from pathlib import Path
import hashlib, json
import pandas as pd

def find_repo_root(start=None):
    p = Path(start or Path.cwd()).resolve()
    for c in [p] + list(p.parents):
        if (c / "saju_engine.py").exists():
            return c
    raise FileNotFoundError("Run inside Chartpalja repo.")

def sha256_file(path):
    h=hashlib.sha256()
    with open(path,"rb") as f:
        for chunk in iter(lambda:f.read(1024*1024),b""):
            h.update(chunk)
    return h.hexdigest()

ROOT=find_repo_root()
FINAL_DIR=ROOT/"research/ml/artifacts/v5_final_roster"
COLL_DIR=ROOT/"research/ml/artifacts/v5_event_collection"
CONTRACT_DIR=ROOT/"research/ml_corpus/v5_ground_truth"
COLL_DIR.mkdir(parents=True,exist_ok=True)

ROSTER=FINAL_DIR/"V5_DEV_SUBJECT_ROSTER_160.csv"
FREEZE=FINAL_DIR/"V5_FINAL_ROSTER_FREEZE_DECISION.json"
CONTRACT=CONTRACT_DIR/"V5_DEV_EVENT_COLLECTION_CONTRACT.json"
WORKLIST=COLL_DIR/"V5_DEV_EVENT_RESEARCH_WORKLIST_8BATCH.csv"
EXPANDED=COLL_DIR/"V5_DEV_EVENT_INTAKE_EXPANDED_12SLOTS.csv"
SUMMARY=COLL_DIR/"V5_DEV_EVENT_BATCH_SUMMARY.csv"

for p in [ROSTER,FREEZE,CONTRACT,WORKLIST,EXPANDED,SUMMARY]:
    if not p.exists(): raise FileNotFoundError(p)

r=pd.read_csv(ROSTER)
f=json.load(open(FREEZE,encoding="utf-8"))
c=json.load(open(CONTRACT,encoding="utf-8"))
w=pd.read_csv(WORKLIST)
e=pd.read_csv(EXPANDED)
s=pd.read_csv(SUMMARY)

assert f["status"]=="V5_FINAL_DEV_AND_CONFIRM_ROSTERS_FROZEN_READY_FOR_DEV_EVENT_COLLECTION"
assert f["dev_roster_sha256"]==sha256_file(ROSTER)
assert c["status"]=="PREDECLARED_BEFORE_V5_DEV_EVENT_RESEARCH"
assert len(r)==160 and r.subject_id.nunique()==160
assert len(w)==160 and w.subject_id.nunique()==160
assert set(r.subject_id)==set(w.subject_id)
assert w.groupby("preassigned_axis").size().to_dict()=={"COMPETITIVE":40,"PROJECT":50,"STATUS":70}
assert w.batch_id.nunique()==8
assert len(e)==160*12
assert e.groupby("subject_id").size().eq(12).all()

print("EVENT COLLECTION FREEZE PREFLIGHT PASS")
display(s)
print("No event research or astrology has been performed by this notebook.")
```
