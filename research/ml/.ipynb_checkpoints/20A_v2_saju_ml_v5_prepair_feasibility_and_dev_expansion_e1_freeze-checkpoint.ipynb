# 20A v2 — V5 Pre-pair Feasibility Gate + DEV Expansion E1 Freeze

This notebook deliberately runs **before any 1–5 year pair-gap inspection**.

The frozen pair gate requires at least:

- total pairable subjects: 80
- COMPETITIVE: 20
- PROJECT: 20
- STATUS: 25

A subject cannot possibly become pairable unless the frozen event corpus already contains
at least one positive and one negative eligible event for that subject. Therefore the
both-polarity subject count is a strict upper bound on later local pairability.

If that necessary condition already fails, this notebook does **not** generate pairs.
Instead, it freezes one new outcome-blind DEV expansion wave E1 using exact-DOB candidates only.

No astrology. No Control. No CONFIRM research.


````python

from pathlib import Path
from datetime import datetime
import hashlib, json, re, unicodedata, ast, time
import numpy as np
import urllib.parse, urllib.request, urllib.error
import pandas as pd
from IPython.display import display

NOTEBOOK_VERSION="SAJU_ML_V5_PREPAIR_FEASIBILITY_EXPANSION_E1_V2_20260817"
SEED=2026081703

def repo_root(start=None):
    p=Path(start or Path.cwd()).resolve()
    for c in [p]+list(p.parents):
        if (c/"saju_engine.py").exists():
            return c
    raise FileNotFoundError("Run inside Chartpalja repo.")

def sha256_file(path):
    h=hashlib.sha256()
    with open(path,"rb") as f:
        for chunk in iter(lambda:f.read(1024*1024),b""):
            h.update(chunk)
    return h.hexdigest()

def norm_name(x):
    s=unicodedata.normalize("NFKD",str(x))
    s="".join(c for c in s if not unicodedata.combining(c))
    s=s.casefold()
    return re.sub(r"[^a-z0-9]+","",s)

def selection_key(axis,qid,name):
    raw=f"{SEED}|{axis}|{qid}|{norm_name(name)}"
    return hashlib.sha256(raw.encode("utf-8")).hexdigest()

ROOT=repo_root()

FREEZE=ROOT/"research/ml/artifacts/v5_event_freeze"
REPAIR=ROOT/"research/ml/artifacts/v5_identity_repair"
AUDIT=ROOT/"research/ml/artifacts/v5_identity_linkage_audit"
CORPUS=ROOT/"research/ml_corpus/v5_ground_truth"

OUT=ROOT/"research/ml/artifacts/v5_dev_expansion_e1"
BATCH_OUT=OUT/"batches"
OUT.mkdir(parents=True,exist_ok=True)
BATCH_OUT.mkdir(parents=True,exist_ok=True)

FREEZE_DEC=FREEZE/"V5_DEV_EVENT_FREEZE_DECISION.json"
FREEZE_MAN=FREEZE/"V5_DEV_EVENT_FREEZE_MANIFEST.json"
SUBCOV=FREEZE/"V5_DEV_EVENT_FREEZE_SUBJECT_COVERAGE.csv"
DEV=REPAIR/"V5_DEV_SUBJECT_ROSTER_160_REPAIRED.csv"
CONF=REPAIR/"V5_CONFIRM_SUBJECT_ROSTER_80_REPAIRED_SEALED.csv"
POOL=AUDIT/"V5_IDENTITY_LINKAGE_AUDIT_CANDIDATE_POOL.csv"
GATE_PROTO=CORPUS/"V5_PREPAIR_FEASIBILITY_GATE_PROTOCOL.json"
EXP_PROTO=CORPUS/"V5_DEV_EXPANSION_WAVE_E1_PROTOCOL.json"
SUPP_PROTO=CORPUS/"V5_DEV_EXPANSION_E1_STATUS_SUPPLEMENT_PROTOCOL.json"

for p in [FREEZE_DEC,FREEZE_MAN,SUBCOV,DEV,CONF,POOL,GATE_PROTO,EXP_PROTO,SUPP_PROTO]:
    if not p.exists(): raise FileNotFoundError(p)

fdec=json.load(open(FREEZE_DEC,encoding="utf-8"))
fman=json.load(open(FREEZE_MAN,encoding="utf-8"))
gproto=json.load(open(GATE_PROTO,encoding="utf-8"))
eproto=json.load(open(EXP_PROTO,encoding="utf-8"))
sproto=json.load(open(SUPP_PROTO,encoding="utf-8"))

assert fdec["status"]=="V5_DEV_EVENT_CORPUS_FROZEN_READY_FOR_LOCAL_PAIRING_NUISANCE_GATE"
assert fman["status"]==fdec["status"]
assert sha256_file(SUBCOV)==fman["subject_coverage_sha256"]
assert gproto["status"]=="PREDECLARED_BEFORE_ANY_1_TO_5_YEAR_PAIR_GAP_INSPECTION"
assert eproto["status"]=="PREDECLARED_AFTER_NECESSARY_CONDITION_FAILURE_BEFORE_EXPANSION_MEMBERSHIP_SELECTION"
assert sproto["status"]=="PREDECLARED_AFTER_STATUS_CANDIDATE_SUPPLY_FAILURE_BEFORE_SUPPLEMENT_QUERY"

sub=pd.read_csv(SUBCOV)
dev=pd.read_csv(DEV)
conf=pd.read_csv(CONF)
pool=pd.read_csv(POOL)

assert len(dev)==160 and len(conf)==80

print("20A PREFLIGHT PASS")
print("No pair file or pair-gap calculation has been loaded.")
````

## 1. Necessary-condition gate only

````python

sub["has_positive"]=sub.positive_events>0
sub["has_negative"]=sub.negative_events>0
sub["both_polarities"]=sub.has_positive & sub.has_negative

axis_both=(
    sub.groupby("preassigned_axis").both_polarities.sum()
    .reindex(["COMPETITIVE","PROJECT","STATUS"],fill_value=0)
    .astype(int)
)
total_both=int(sub.both_polarities.sum())

threshold_total=int(gproto["frozen_thresholds"]["min_pairable_subjects_total"])
threshold_axis=gproto["frozen_thresholds"]["min_pairable_subjects_by_axis"]

feasibility=pd.DataFrame([
    {
        "scope":"TOTAL",
        "both_polarity_upper_bound":total_both,
        "frozen_min_pairable":threshold_total,
        "necessary_condition_pass":total_both>=threshold_total
    },
    *[
        {
            "scope":axis,
            "both_polarity_upper_bound":int(axis_both[axis]),
            "frozen_min_pairable":int(threshold_axis[axis]),
            "necessary_condition_pass":int(axis_both[axis])>=int(threshold_axis[axis])
        }
        for axis in ["COMPETITIVE","PROJECT","STATUS"]
    ]
])

display(feasibility)

necessary_pass=bool(feasibility.necessary_condition_pass.all())
if necessary_pass:
    status="V5_PREPAIR_FEASIBILITY_NECESSARY_CONDITION_PASS_PROCEED_TO_PAIR_GENERATION"
else:
    status="V5_PREPAIR_FEASIBILITY_FAIL_EXPANSION_REQUIRED"

feas_path=OUT/"V5_PREPAIR_FEASIBILITY_SUMMARY.csv"
feasibility.to_csv(feas_path,index=False)

gate_decision={
    "version":"V5_PREPAIR_FEASIBILITY_DECISION_V1",
    "created_at":datetime.now().isoformat(timespec="seconds"),
    "status":status,
    "both_polarity_upper_bound_total":total_both,
    "both_polarity_upper_bound_by_axis":axis_both.to_dict(),
    "frozen_min_pairable_total":threshold_total,
    "frozen_min_pairable_by_axis":threshold_axis,
    "pair_gap_inspected":False,
    "pairs_generated":False,
    "astrology_generated":False,
    "control_scored":False,
    "confirm_researched":False,
    "feasibility_summary_sha256":sha256_file(feas_path),
    "next_rule":(
        "Freeze and collect a whole new DEV expansion wave under V5_DEV_EXPANSION_WAVE_E1_PROTOCOL."
        if not necessary_pass else
        "Proceed to deterministic local 1-5y pair generation."
    )
}
gate_dec_path=OUT/"V5_PREPAIR_FEASIBILITY_DECISION.json"
json.dump(gate_decision,open(gate_dec_path,"w",encoding="utf-8"),ensure_ascii=False,indent=2)

print(json.dumps(gate_decision,ensure_ascii=False,indent=2))

# This notebook version is intended for the currently observed necessary-condition failure.
assert status=="V5_PREPAIR_FEASIBILITY_FAIL_EXPANSION_REQUIRED", (
    "Necessary condition unexpectedly passed. Stop here and build the pair-generation notebook instead."
)
````

## 2. Freeze one whole new exact-DOB DEV expansion wave E1

````python

# Candidate identities from the audited pool must already have exact day-level DOB linkage verification.
cand=pool[
    (pool.linkage_status=="PASS_EXACT_DOB")
    & pool.wikidata_id.notna()
].copy()

cand["norm_name_e1"]=cand.name.map(norm_name)

selected_names=set(dev.name.map(norm_name)) | set(conf.name.map(norm_name))
selected_qids=set(
    pd.concat([dev.wikidata_id,conf.wikidata_id])
    .dropna().astype(str)
)

cand=cand[
    ~cand.norm_name_e1.isin(selected_names)
    & ~cand.wikidata_id.astype(str).isin(selected_qids)
].copy()

# -------------------------------------------------------------------
# E1 STATUS supply supplement
# -------------------------------------------------------------------
# The original audited candidate pool has only 36 unused exact-DOB STATUS identities,
# below the already frozen E1 STATUS quota of 70.
#
# Do NOT lower the quota. Instead, under the separately frozen supplement protocol,
# generate additional STATUS candidates from role families selected solely for
# formal institutional/organizational authority relevance.
#
# No event outcomes, chronology, pairability, astrology, or Control are queried.
# New identities are admitted only when:
#   1) Wikidata English label matches a unique strict Rodden-AA birth-snapshot name, AND
#   2) Wikidata P569 date exactly equals the Rodden-AA calendar birth date.
# -------------------------------------------------------------------

quotas=eproto["wave"]["axis_quotas"]

def current_supply(df):
    return df.axis.value_counts().to_dict()

availability=current_supply(cand)
status_need=max(0,int(quotas["STATUS"])-int(availability.get("STATUS",0)))

supplement_used=False
status_supplement_pool_path=OUT/"V5_DEV_EXPANSION_E1_STATUS_SUPPLEMENT_EXACT_DOB_POOL.csv"
status_supplement_query_manifest_path=OUT/"V5_DEV_EXPANSION_E1_STATUS_SUPPLEMENT_QUERY_MANIFEST.json"

if status_need > 0:
    print("STATUS exact-DOB supply shortage:", availability.get("STATUS",0), "/", quotas["STATUS"])
    print("Running frozen role-only STATUS supplement. Minimum extra needed:", status_need)

    supplement_used=True

    BIRTH_PATH=ROOT/"research/ml/artifacts/v4_unified_dev_roster/PersonList-15k.csv"
    if not BIRTH_PATH.exists():
        raise FileNotFoundError(BIRTH_PATH)

    birth=pd.read_csv(BIRTH_PATH)
    required_birth_cols={"RowKey","BirthTime","Gender","Name","Notes"}
    if not required_birth_cols.issubset(set(birth.columns)):
        raise RuntimeError("Unexpected PersonList-15k schema: %s" % list(birth.columns))

    def parse_birth_blob(blob):
        obj=json.loads(str(blob))
        std=obj["StdTime"]
        loc=obj["Location"]
        m=re.match(
            r"^(\d{1,2}):(\d{2})\s+(\d{2})/(\d{2})/(\d{4})\s+([+-]\d{2}:\d{2})$",
            std.strip()
        )
        if not m:
            raise ValueError(std)
        hh,mm,dd,mo,yyyy,offset=m.groups()
        return {
            "birth_date":"%04d-%02d-%02d" % (int(yyyy),int(mo),int(dd)),
            "birth_time":"%02d:%02d" % (int(hh),int(mm)),
            "utc_offset":offset,
            "birth_year":int(yyyy),
            "birth_place":loc["Name"],
            "longitude":float(loc["Longitude"]),
            "latitude":float(loc["Latitude"]),
        }

    def rodden_grade(notes):
        s=str(notes)
        try:
            obj=ast.literal_eval(s)
            if isinstance(obj,dict):
                return str(obj.get("rodden") or "").strip().upper()
        except Exception:
            pass
        m=re.search(r"rodden['\"]?\s*:\s*['\"]([^'\"]+)['\"]",s,flags=re.I)
        return m.group(1).strip().upper() if m else ""

    aa_rows=[]
    for _,r in birth.iterrows():
        if rodden_grade(r["Notes"])!="AA":
            continue
        try:
            p=parse_birth_blob(r["BirthTime"])
        except Exception:
            continue
        if not (1900 <= int(p["birth_year"]) <= 1995):
            continue
        aa_rows.append({
            "source_row_key":r["RowKey"],
            "source_name":r["Name"],
            "gender":str(r["Gender"]).lower(),
            "norm_name_e1":norm_name(r["Name"]),
            **p,
        })

    aa=pd.DataFrame(aa_rows)
    # Same-name AA rows are unsafe for name linkage and are excluded.
    name_n=aa.groupby("norm_name_e1").size()
    ambiguous_names=set(name_n[name_n>1].index)
    aa_unique=aa[~aa.norm_name_e1.isin(ambiguous_names)].copy()

    if len(aa_unique)<500:
        raise RuntimeError("Unexpectedly small strict-AA unique birth snapshot.")

    STATUS_SUPPLEMENT_ROLE_QIDS=sproto["status_supplement_role_qids"]
    WIKIDATA_ENDPOINT="https://query.wikidata.org/sparql"
    QUERY_CACHE=OUT/"status_supplement_query_cache"
    QUERY_CACHE.mkdir(parents=True,exist_ok=True)

    def wikidata_role_dob_query(qid,limit=5000):
        # Lightweight role query: occupation + English label + date of birth only.
        # P569 is used solely for exact identity linkage.
        return """
PREFIX wd: <http://www.wikidata.org/entity/>
PREFIX wdt: <http://www.wikidata.org/prop/direct/>
PREFIX rdfs: <http://www.w3.org/2000/01/rdf-schema#>

SELECT DISTINCT ?person ?personLabel ?dob WHERE {
  ?person wdt:P106 wd:%s ;
          wdt:P569 ?dob ;
          rdfs:label ?personLabel .
  FILTER(LANG(?personLabel) = "en")
  FILTER(YEAR(?dob) >= 1900 && YEAR(?dob) <= 1995)
}
LIMIT %d
""" % (qid,int(limit))

    def run_wdqs(query,cache_path,max_attempts=4):
        if cache_path.exists():
            return json.load(open(cache_path,encoding="utf-8"))

        params=urllib.parse.urlencode({"query":query,"format":"json"})
        url=WIKIDATA_ENDPOINT+"?"+params
        headers={
            "Accept":"application/sparql-results+json",
            "User-Agent":"Chartpalja-Saju-Research/1.0 (E1 role-only exact-DOB candidate supply)"
        }
        last=None
        for attempt in range(max_attempts):
            try:
                req=urllib.request.Request(url,headers=headers)
                with urllib.request.urlopen(req,timeout=90) as resp:
                    obj=json.loads(resp.read().decode("utf-8"))
                json.dump(obj,open(cache_path,"w",encoding="utf-8"),ensure_ascii=False)
                return obj
            except Exception as e:
                last=e
                if attempt+1<max_attempts:
                    sleep_s=3*(2**attempt)
                    print("WDQS retry",sleep_s,"sec:",repr(e))
                    time.sleep(sleep_s)
        print("ROLE QUERY FAILED:",cache_path.name,repr(last))
        return {"results":{"bindings":[]}}

    supp_rows=[]
    query_manifest={}
    for role_name,qid in STATUS_SUPPLEMENT_ROLE_QIDS.items():
        cache_path=QUERY_CACHE/f"STATUS__{role_name}.json"
        print("supplement query:",role_name,qid)
        obj=run_wdqs(wikidata_role_dob_query(qid),cache_path)
        bindings=obj.get("results",{}).get("bindings",[])
        query_manifest[role_name]={
            "qid":qid,
            "returned_rows":len(bindings),
            "cache_file":cache_path.name,
            "cache_sha256":sha256_file(cache_path) if cache_path.exists() else None,
        }
        print("  returned:",len(bindings))

        for b in bindings:
            uri=(b.get("person") or {}).get("value","")
            label=(b.get("personLabel") or {}).get("value","")
            dob=(b.get("dob") or {}).get("value","")
            qid_person=uri.rsplit("/",1)[-1] if uri else ""
            m=re.match(r"^(\d{4})-(\d{2})-(\d{2})",dob)
            if not (qid_person and label and m):
                continue
            yyyy,mo,dd=map(int,m.groups())
            # Reject low-precision/unknown Wikidata dates such as month/day 00.
            if mo<1 or dd<1:
                continue
            birth_date="%04d-%02d-%02d"%(yyyy,mo,dd)
            supp_rows.append({
                "axis":"STATUS",
                "role_family":role_name,
                "role_qid":qid,
                "wikidata_id":qid_person,
                "name":label,
                "norm_name_e1":norm_name(label),
                "wikidata_birth_date":birth_date,
            })

    role_supp=pd.DataFrame(supp_rows)
    if len(role_supp)==0:
        raise RuntimeError("STATUS supplement queries returned no usable role+DOB rows.")

    # One candidate identity can have several supplemental occupations.
    role_supp=(
        role_supp.sort_values(["wikidata_id","role_family"])
        .drop_duplicates(["wikidata_id","norm_name_e1","wikidata_birth_date"])
        .reset_index(drop=True)
    )

    # Exact name + exact DOB linkage to the same strict Rodden-AA snapshot.
    resolved=role_supp.merge(
        aa_unique,
        left_on=["norm_name_e1","wikidata_birth_date"],
        right_on=["norm_name_e1","birth_date"],
        how="inner",
        validate="many_to_one"
    )

    # Exclude all currently selected DEV/CONFIRM identities.
    resolved=resolved[
        ~resolved.norm_name_e1.isin(selected_names)
        & ~resolved.wikidata_id.astype(str).isin(selected_qids)
    ].copy()

    # Prevent unexpected one-QID/multiple-AA or duplicate-normalized identity issues.
    resolved=(
        resolved.sort_values(["wikidata_id","role_family","source_row_key"])
        .drop_duplicates("wikidata_id")
        .drop_duplicates("norm_name_e1")
        .reset_index(drop=True)
    )

    resolved["linkage_status"]="PASS_EXACT_DOB"
    resolved["candidate_source"]="V5_E1_STATUS_ROLE_SUPPLEMENT_EXACT_DOB"
    resolved["rodden_rating"]="AA"
    resolved["birth_source"]="Frozen VedAstro PersonList-15k snapshot"
    resolved["selection_information_used"]="ROLE_ONLY + WIKIDATA_P569_EXACT_DOB_LINKAGE"
    resolved["event_collection_started"]=False
    resolved["astrology_scored"]=False

    # Match candidate-pool schema by adding absent columns.
    for c in cand.columns:
        if c not in resolved.columns:
            resolved[c]=np.nan
    for c in resolved.columns:
        if c not in cand.columns:
            cand[c]=np.nan

    # Existing independently audited pool gets first priority if the same person appears.
    cand["_source_priority"]=0
    resolved["_source_priority"]=1

    cand=pd.concat(
        [cand[resolved.columns],resolved],
        ignore_index=True,
        sort=False
    )

    cand=(
        cand.sort_values(
            ["axis","_source_priority","wikidata_id","norm_name_e1","role_family"],
            kind="stable"
        )
        .drop_duplicates("wikidata_id")
        .drop_duplicates(["axis","norm_name_e1"])
        .reset_index(drop=True)
    )
    cand=cand.drop(columns=["_source_priority"],errors="ignore")

    resolved.to_csv(status_supplement_pool_path,index=False)
    json.dump(
        {
            "version":"V5_E1_STATUS_SUPPLEMENT_QUERY_MANIFEST_V1",
            "status":"STATUS_SUPPLEMENT_QUERIED_EXACT_DOB_LINKED",
            "role_queries":query_manifest,
            "strict_AA_unique_birth_rows":int(len(aa_unique)),
            "resolved_exact_DOB_status_candidates_n":int(len(resolved)),
            "birth_snapshot_sha256":sha256_file(BIRTH_PATH),
            "supplement_pool_sha256":sha256_file(status_supplement_pool_path),
            "events_used":False,
            "pairability_used":False,
            "chronology_used":False,
            "astrology_used":False,
            "control_used":False,
        },
        open(status_supplement_query_manifest_path,"w",encoding="utf-8"),
        ensure_ascii=False,
        indent=2
    )

# Selection key is applied only after all permitted candidate supply is frozen.
cand["selection_key_e1"]=cand.apply(
    lambda r:selection_key(str(r.axis),str(r.wikidata_id),r["name"]),
    axis=1
)

cand=(
    cand.sort_values(["axis","selection_key_e1","wikidata_id","norm_name_e1"])
    .drop_duplicates("wikidata_id")
    .drop_duplicates(["axis","norm_name_e1"])
    .reset_index(drop=True)
)

availability=cand.axis.value_counts().to_dict()

suff=pd.DataFrame([
    {
        "axis":axis,
        "quota":int(quotas[axis]),
        "available_exact_DOB_candidates":int(availability.get(axis,0)),
        "sufficient":int(availability.get(axis,0))>=int(quotas[axis])
    }
    for axis in ["COMPETITIVE","PROJECT","STATUS"]
])
display(suff)

supply_diag_path=OUT/"V5_DEV_EXPANSION_E1_CANDIDATE_SUPPLY_AFTER_STATUS_SUPPLEMENT.csv"
suff.to_csv(supply_diag_path,index=False)

if not suff.sufficient.all():
    raise RuntimeError(
        "Exact-DOB supply still insufficient after the frozen STATUS role supplement. "
        "Do not lower quotas. Send V5_DEV_EXPANSION_E1_CANDIDATE_SUPPLY_AFTER_STATUS_SUPPLEMENT.csv."
    )

chosen_parts=[]
for axis in ["COMPETITIVE","PROJECT","STATUS"]:
    x=(
        cand[cand.axis==axis]
        .sort_values(["selection_key_e1","wikidata_id","norm_name_e1"])
        .head(int(quotas[axis]))
        .copy()
    )
    assert len(x)==int(quotas[axis])
    chosen_parts.append(x)

chosen=pd.concat(chosen_parts,ignore_index=True)

assert len(chosen)==160
assert chosen.wikidata_id.nunique()==160
assert chosen.norm_name_e1.nunique()==160
assert set(chosen.wikidata_id.astype(str)).isdisjoint(selected_qids)

print("E1 candidate supply PASS")
print("supplement_used:",supplement_used)
if supplement_used:
    print("status supplement exact-DOB candidates:", len(pd.read_csv(status_supplement_pool_path)))
````

## 3. Assign fixed subject IDs and the same 8-batch axis schedule

````python

# Same batch composition as original 160-person DEV.
batch_specs={
    1:{"COMPETITIVE":5,"PROJECT":7,"STATUS":9},
    2:{"COMPETITIVE":5,"PROJECT":7,"STATUS":9},
    3:{"COMPETITIVE":5,"PROJECT":6,"STATUS":9},
    4:{"COMPETITIVE":5,"PROJECT":6,"STATUS":9},
    5:{"COMPETITIVE":5,"PROJECT":6,"STATUS":9},
    6:{"COMPETITIVE":5,"PROJECT":6,"STATUS":9},
    7:{"COMPETITIVE":5,"PROJECT":6,"STATUS":8},
    8:{"COMPETITIVE":5,"PROJECT":6,"STATUS":8},
}
assert sum(v["COMPETITIVE"] for v in batch_specs.values())==40
assert sum(v["PROJECT"] for v in batch_specs.values())==50
assert sum(v["STATUS"] for v in batch_specs.values())==70

queues={
    axis:chosen[chosen.axis==axis].sort_values(
        ["selection_key_e1","wikidata_id"]
    ).reset_index(drop=True)
    for axis in ["COMPETITIVE","PROJECT","STATUS"]
}
ptr={k:0 for k in queues}

schedule=[]
seq=1
for b in range(1,9):
    order=1
    # Preserve a deterministic mixed-axis order instead of grouping an entire batch by axis.
    remaining=batch_specs[b].copy()
    axis_cycle=["COMPETITIVE","PROJECT","STATUS"]
    while sum(remaining.values())>0:
        for axis in axis_cycle:
            if remaining[axis]<=0:
                continue
            r=queues[axis].iloc[ptr[axis]]
            ptr[axis]+=1
            remaining[axis]-=1

            rec=r.to_dict()
            rec.update({
                "subject_id":f"V5DEVX_{seq:03d}",
                "split":"DEV_EXPANSION_E1",
                "preassigned_axis":axis,
                "expansion_wave":"E1",
                "batch_id":b,
                "research_order_in_batch":order,
                "event_collection_started":False,
                "astrology_scored":False,
            })
            schedule.append(rec)
            seq+=1
            order+=1

roster=pd.DataFrame(schedule)

assert len(roster)==160
assert roster.subject_id.nunique()==160
assert roster.preassigned_axis.value_counts().to_dict()=={
    "STATUS":70,"PROJECT":50,"COMPETITIVE":40
}
assert roster.batch_id.value_counts().sort_index().to_dict()=={
    1:21,2:21,3:20,4:20,5:20,6:20,7:19,8:19
}

# Remove helper-only columns from published roster.
for c in ["norm_name_e1"]:
    if c in roster.columns:
        roster=roster.drop(columns=c)

roster_path=OUT/"V5_DEV_EXPANSION_E1_ROSTER_160.csv"
roster.to_csv(roster_path,index=False)

work_cols=[
    "batch_id","research_order_in_batch","subject_id","name",
    "preassigned_axis","candidate_source","birth_date","birth_place",
    "gender","wikidata_id","rodden_rating","source_row_key"
]
work_cols=[c for c in work_cols if c in roster.columns]
work=roster[work_cols].copy()

work_path=OUT/"V5_DEV_EXPANSION_E1_EVENT_WORKLIST_8BATCH.csv"
work.to_csv(work_path,index=False)

for b in range(1,9):
    wb=work[work.batch_id==b].sort_values("research_order_in_batch").copy()
    wb.to_csv(
        BATCH_OUT/f"V5_DEV_EXPANSION_E1_BATCH_{b:02d}_SUBJECTS.csv",
        index=False
    )

print("Expansion E1 roster frozen:",len(roster))
display(
    work.groupby(["batch_id","preassigned_axis"]).size().unstack(fill_value=0)
)
````

## 4. Freeze decision
