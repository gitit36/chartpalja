# 18C — V5 Post-Repair Validation + Batch 1–3 Sanitization

This notebook runs **after 18B identity repair**.

It does two things only:

1. Re-validates that every selected DEV/CONFIRM identity now satisfies the frozen `PASS_EXACT_DOB` rule.
2. Removes old Batch 1–3 event rows only for `subject_id` slots whose identity changed.

It does **not**:
- change membership again,
- use event polarity or pairability to decide what survives,
- generate astrology,
- inspect Control,
- resume Batch 4.

After this notebook, only the fixed repaired slots emitted in
`V5_BATCH_01_03_RERESEARCH_SUBJECTS.csv` may be researched.


````python

from pathlib import Path
from datetime import datetime
import hashlib, json, re, unicodedata
import numpy as np
import pandas as pd
from IPython.display import display

NOTEBOOK_VERSION="SAJU_ML_V5_POST_REPAIR_VALIDATION_20260817"
KEEP_STATUS="PASS_EXACT_DOB"

def find_repo_root(start=None):
    p=Path(start or Path.cwd()).resolve()
    for c in [p]+list(p.parents):
        if (c/"saju_engine.py").exists():
            return c
    raise FileNotFoundError("Run inside Chartpalja repo.")

def sha256_file(path):
    h=hashlib.sha256()
    with open(path,"rb") as f:
        for chunk in iter(lambda:f.read(1024*1024),b""):
            h.update(chunk)
    return h.hexdigest()

def norm_name(x):
    s=unicodedata.normalize("NFKD",str(x))
    s="".join(c for c in s if not unicodedata.combining(c))
    s=s.casefold()
    return re.sub(r"[^a-z0-9]+","",s)

ROOT=find_repo_root()

REPAIR=ROOT/"research/ml/artifacts/v5_identity_repair"
AUDIT=ROOT/"research/ml/artifacts/v5_identity_linkage_audit"
OLD_EVENT=ROOT/"research/ml/artifacts/v5_event_collection"
NEW_EVENT=ROOT/"research/ml/artifacts/v5_event_collection_repaired"
CORPUS=ROOT/"research/ml_corpus/v5_ground_truth"

OUT=ROOT/"research/ml/artifacts/v5_post_repair_validation"
SAN=NEW_EVENT/"results_sanitized"
OUT.mkdir(parents=True,exist_ok=True)
SAN.mkdir(parents=True,exist_ok=True)

DEV_PATH=REPAIR/"V5_DEV_SUBJECT_ROSTER_160_REPAIRED.csv"
CONFIRM_PATH=REPAIR/"V5_CONFIRM_SUBJECT_ROSTER_80_REPAIRED_SEALED.csv"
DEV_MAP_PATH=REPAIR/"V5_IDENTITY_REPAIR_MAP_DEV.csv"
CONFIRM_MAP_PATH=REPAIR/"V5_IDENTITY_REPAIR_MAP_CONFIRM_SEALED.csv"
REPAIR_DECISION_PATH=REPAIR/"V5_IDENTITY_REPAIR_DECISION.json"

DEV_AUDIT_PATH=AUDIT/"V5_IDENTITY_LINKAGE_AUDIT_DEV.csv"
CONFIRM_AUDIT_PATH=AUDIT/"V5_IDENTITY_LINKAGE_AUDIT_CONFIRM_SEALED.csv"
POOL_AUDIT_PATH=AUDIT/"V5_IDENTITY_LINKAGE_AUDIT_CANDIDATE_POOL.csv"

INVALID_PATH=NEW_EVENT/"V5_EVENT_RESEARCH_INVALIDATED_SLOTS.csv"
REPAIRED_WORKLIST_PATH=NEW_EVENT/"V5_DEV_EVENT_RESEARCH_WORKLIST_8BATCH_REPAIRED.csv"
PROTOCOL_PATH=CORPUS/"V5_POST_REPAIR_VALIDATION_PROTOCOL.json"

for p in [
    DEV_PATH,CONFIRM_PATH,DEV_MAP_PATH,CONFIRM_MAP_PATH,REPAIR_DECISION_PATH,
    DEV_AUDIT_PATH,CONFIRM_AUDIT_PATH,POOL_AUDIT_PATH,
    INVALID_PATH,REPAIRED_WORKLIST_PATH,PROTOCOL_PATH
]:
    if not p.exists():
        raise FileNotFoundError(p)

dev=pd.read_csv(DEV_PATH)
confirm=pd.read_csv(CONFIRM_PATH)
dmap=pd.read_csv(DEV_MAP_PATH)
cmap=pd.read_csv(CONFIRM_MAP_PATH)
adev=pd.read_csv(DEV_AUDIT_PATH)
aconfirm=pd.read_csv(CONFIRM_AUDIT_PATH)
apool=pd.read_csv(POOL_AUDIT_PATH)
invalid=pd.read_csv(INVALID_PATH)
worklist=pd.read_csv(REPAIRED_WORKLIST_PATH)

repair=json.load(open(REPAIR_DECISION_PATH,encoding="utf-8"))
protocol=json.load(open(PROTOCOL_PATH,encoding="utf-8"))

assert repair["status"]=="V5_IDENTITY_REPAIR_APPLIED_READY_FOR_POST_REPAIR_VALIDATION"
assert sha256_file(DEV_PATH)==repair["repaired_DEV_sha256"]
assert sha256_file(CONFIRM_PATH)==repair["repaired_CONFIRM_sha256"]
assert sha256_file(DEV_MAP_PATH)==repair["DEV_repair_map_sha256"]
assert sha256_file(CONFIRM_MAP_PATH)==repair["CONFIRM_repair_map_sha256"]
assert sha256_file(REPAIRED_WORKLIST_PATH)==repair["repaired_worklist_sha256"]
assert sha256_file(INVALID_PATH)==repair["invalidated_slots_sha256"]
assert protocol["status"]=="PREDECLARED_AFTER_IDENTITY_REPAIR_BEFORE_EVENT_RESEARCH_RESUMPTION"

assert len(dev)==160 and dev.subject_id.nunique()==160
assert len(confirm)==80 and confirm.subject_id.nunique()==80
assert len(worklist)==160 and set(worklist.subject_id)==set(dev.subject_id)
assert len(dmap)==repair["DEV_replaced_n"]
assert len(cmap)==repair["CONFIRM_replaced_n"]

print("18C PREFLIGHT PASS")
````

## 1. Reconstruct exact-DOB validation state for every selected slot

````python

old_dev_status=adev.set_index("subject_id")["linkage_status"].to_dict()
old_confirm_status=aconfirm.set_index("subject_id")["linkage_status"].to_dict()

exact_qids=set(
    apool.loc[
        apool.linkage_status==KEEP_STATUS,
        "wikidata_id"
    ].dropna().astype(str)
)

dmap_by_sid=dmap.set_index("subject_id")
cmap_by_sid=cmap.set_index("subject_id")

validation_rows=[]

for _,r in dev.iterrows():
    sid=r.subject_id
    repaired=sid in set(dmap.subject_id)
    if repaired:
        evidence_status=str(dmap_by_sid.loc[sid,"new_linkage_status"])
        evidence_qid=str(dmap_by_sid.loc[sid,"new_wikidata_id"])
        valid=(evidence_status==KEEP_STATUS and evidence_qid in exact_qids)
    else:
        evidence_status=str(old_dev_status[sid])
        evidence_qid=str(r.wikidata_id)
        valid=(evidence_status==KEEP_STATUS)

    validation_rows.append({
        "split":"DEV",
        "subject_id":sid,
        "name":r["name"],
        "preassigned_axis":r.preassigned_axis,
        "was_repaired":repaired,
        "identity_evidence_status":evidence_status,
        "wikidata_id":evidence_qid,
        "exact_DOB_validated":bool(valid)
    })

# CONFIRM remains aggregate/sealed; full validation rows never exported in user-return files.
confirm_valid=[]
for _,r in confirm.iterrows():
    sid=r.subject_id
    repaired=sid in set(cmap.subject_id)
    if repaired:
        st=str(cmap_by_sid.loc[sid,"new_linkage_status"])
        qid=str(cmap_by_sid.loc[sid,"new_wikidata_id"])
        valid=(st==KEEP_STATUS and qid in exact_qids)
    else:
        st=str(old_confirm_status[sid])
        valid=(st==KEEP_STATUS)
    confirm_valid.append(bool(valid))

vdev=pd.DataFrame(validation_rows)

assert vdev.exact_DOB_validated.all()
assert all(confirm_valid)
assert set(dev.name.map(norm_name)).isdisjoint(set(confirm.name.map(norm_name)))
assert set(dev.wikidata_id.astype(str)).isdisjoint(set(confirm.wikidata_id.astype(str)))
assert dev.preassigned_axis.value_counts().to_dict()=={
    "STATUS":70,"PROJECT":50,"COMPETITIVE":40
}
assert confirm.preassigned_axis.value_counts().to_dict()=={
    "STATUS":35,"PROJECT":25,"COMPETITIVE":20
}

vdev_path=OUT/"V5_POST_REPAIR_IDENTITY_VALIDATION_DEV.csv"
vdev.to_csv(vdev_path,index=False)

print("All 160 DEV and all 80 sealed CONFIRM slots satisfy PASS_EXACT_DOB.")
print("DEV repaired:",int(vdev.was_repaired.sum()))
````

## 2. Determine exactly which prior Batch 1–3 slots are invalidated

````python

prior_invalid=invalid[invalid.batch_id.isin([1,2,3])].copy()
prior_invalid=prior_invalid.sort_values(
    ["batch_id","research_order_in_batch","subject_id"]
).reset_index(drop=True)

# Repaired identities and scheduling must match repaired worklist.
rr=prior_invalid[[
    "subject_id","batch_id","research_order_in_batch",
    "axis","old_name","new_name","new_birth_date","new_birth_place","new_wikidata_id"
]].merge(
    worklist[[
        "subject_id","name","preassigned_axis","birth_date","birth_place","candidate_source"
    ]],
    on="subject_id",
    how="left",
    validate="one_to_one",
    suffixes=("_repair_map","_worklist")
)

assert rr.name.eq(rr.new_name).all()
assert rr.preassigned_axis.eq(rr.axis).all()
assert rr.birth_date.astype(str).eq(rr.new_birth_date.astype(str)).all()
assert len(rr)>0

rr["research_status"]="RERESEARCH_REQUIRED_REPAIRED_IDENTITY"
rr["old_event_rows_must_be_used"]=False

rr_path=OUT/"V5_BATCH_01_03_RERESEARCH_SUBJECTS.csv"
rr.to_csv(rr_path,index=False)

print("Previously researched slots requiring re-research:",len(rr))
display(rr[[
    "batch_id","research_order_in_batch","subject_id",
    "axis","old_name","new_name"
]])
````

## 3. Sanitize Batch 1–3 outputs by subject_id only

````python

batch_summaries=[]

for b in [1,2,3]:
    src=OLD_EVENT/"results"/f"batch_{b:02d}"
    dst=SAN/f"batch_{b:02d}"
    dst.mkdir(parents=True,exist_ok=True)

    events_path=src/f"V5_DEV_EVENT_BATCH_{b:02d}_EVENTS.csv"
    audit_path=src/f"V5_DEV_EVENT_BATCH_{b:02d}_SUBJECT_SWEEP_AUDIT.csv"
    manifest_path=src/f"V5_DEV_EVENT_BATCH_{b:02d}_RESEARCH_MANIFEST.json"

    for p in [events_path,audit_path,manifest_path]:
        if not p.exists():
            raise FileNotFoundError(
                f"Missing original Batch {b} research output: {p}"
            )

    ev=pd.read_csv(events_path)
    au=pd.read_csv(audit_path)
    old_manifest=json.load(open(manifest_path,encoding="utf-8"))

    bad_ids=set(
        prior_invalid.loc[
            prior_invalid.batch_id==b,
            "subject_id"
        ].astype(str)
    )

    # Critical guardrail: sanitization depends ONLY on fixed invalidated subject IDs.
    ev_keep=ev[~ev.subject_id.astype(str).isin(bad_ids)].copy()
    au_keep=au[~au.subject_id.astype(str).isin(bad_ids)].copy()

    removed_event_rows=int(len(ev)-len(ev_keep))
    removed_audit_rows=int(len(au)-len(au_keep))

    # Ensure no unaffected subject row was altered.
    assert set(ev_keep.subject_id.astype(str)).isdisjoint(bad_ids)
    assert set(au_keep.subject_id.astype(str)).isdisjoint(bad_ids)

    ev_out=dst/f"V5_DEV_EVENT_BATCH_{b:02d}_EVENTS_SANITIZED.csv"
    au_out=dst/f"V5_DEV_EVENT_BATCH_{b:02d}_SUBJECT_SWEEP_AUDIT_SANITIZED.csv"
    ev_keep.to_csv(ev_out,index=False)
    au_keep.to_csv(au_out,index=False)

    sm={
        "version":f"V5_DEV_EVENT_BATCH_{b:02d}_SANITIZED_V1",
        "status":"SANITIZED_AWAITING_RESEARCH_FOR_REPAIRED_SLOTS"
                 if bad_ids else
                 "SANITIZED_COMPLETE_NO_REPAIRED_SLOT_IN_BATCH",
        "batch_id":b,
        "original_manifest_status":old_manifest.get("status"),
        "original_subjects":int(old_manifest.get("n_subjects",au.subject_id.nunique())),
        "invalidated_subject_ids":sorted(bad_ids),
        "invalidated_subject_n":len(bad_ids),
        "removed_event_rows":removed_event_rows,
        "removed_audit_rows":removed_audit_rows,
        "retained_event_rows":int(len(ev_keep)),
        "retained_subject_audit_rows":int(len(au_keep)),
        "events_sanitized_sha256":sha256_file(ev_out),
        "audit_sanitized_sha256":sha256_file(au_out),
        "sanitization_rule":"subject_id invalidation only; event content/polarity not consulted",
        "ready_for_final_batch_freeze":False
    }
    sm_path=dst/f"V5_DEV_EVENT_BATCH_{b:02d}_SANITIZATION_MANIFEST.json"
    json.dump(sm,open(sm_path,"w",encoding="utf-8"),ensure_ascii=False,indent=2)
    batch_summaries.append(sm)

summary_df=pd.DataFrame(batch_summaries)
summary_path=OUT/"V5_BATCH_01_03_SANITIZATION_SUMMARY.csv"
summary_df.to_csv(summary_path,index=False)

display(summary_df[[
    "batch_id","invalidated_subject_n",
    "removed_event_rows","retained_event_rows","retained_subject_audit_rows"
]])
````

## 4. Freeze post-repair validation state

````python

# Validate repaired Batch 04 file exists, but do not authorize its research yet.
batch4_path=NEW_EVENT/"batches"/"V5_DEV_EVENT_BATCH_04_SUBJECTS_REPAIRED.csv"
if not batch4_path.exists():
    raise FileNotFoundError(batch4_path)

b4=pd.read_csv(batch4_path)
assert len(b4)==20
assert set(b4.subject_id)==set(worklist.loc[worklist.batch_id==4,"subject_id"])

decision={
    "version":"V5_POST_REPAIR_VALIDATION_DECISION_V1",
    "notebook_version":NOTEBOOK_VERSION,
    "created_at":datetime.now().isoformat(timespec="seconds"),
    "status":"V5_POST_REPAIR_VALIDATION_PASS_RERESEARCH_6_REQUIRED",
    "DEV_exact_DOB_validated_n":160,
    "CONFIRM_exact_DOB_validated_n":80,
    "DEV_repaired_n":int(len(dmap)),
    "CONFIRM_repaired_n":int(len(cmap)),
    "previously_researched_repaired_slots_n":int(len(rr)),
    "previously_researched_repaired_slots_by_batch":
        rr.batch_id.value_counts().sort_index().to_dict(),
    "reresearch_subjects_sha256":sha256_file(rr_path),
    "sanitization_summary_sha256":sha256_file(summary_path),
    "DEV_validation_sha256":sha256_file(vdev_path),
    "repaired_batch04_subjects_sha256":sha256_file(batch4_path),
    "rules":{
        "membership_changed_in_18C":False,
        "event_content_used_for_sanitization":False,
        "event_polarity_used_for_sanitization":False,
        "pairability_used":False,
        "chronology_used":False,
        "astrology_used":False,
        "control_used":False,
        "confirm_details_remain_sealed":True
    },
    "event_research_may_resume_for_reresearch_6":True,
    "batch_04_may_resume":False,
    "next_rule":(
        "Research only the subjects in V5_BATCH_01_03_RERESEARCH_SUBJECTS.csv. "
        "Merge those new rows with sanitized Batch 1-3 outputs and validate completion "
        "before resuming repaired Batch 04."
    )
}

decision_path=OUT/"V5_POST_REPAIR_VALIDATION_DECISION.json"
json.dump(
    decision,
    open(decision_path,"w",encoding="utf-8"),
    ensure_ascii=False,
    indent=2
)

print(json.dumps(decision,ensure_ascii=False,indent=2))
````
