# 10 — V4 Unified DEV Wave 2 Freeze + Combined QA

**Purpose:** freeze/QA Wave 2 event labels, preserve all 44 frozen members, combine frozen Wave 1 + Wave 2 usable pairs, and determine whether axis targets are met.

**Hard rule:** this notebook does **not** generate astrology features or score any astrology model.


```python

from pathlib import Path
from datetime import datetime
import hashlib, json
import numpy as np
import pandas as pd

NOTEBOOK_VERSION = "SAJU_ML_V4_UNIFIED_DEV_WAVE2_FREEZE_COMBINED_QA_20260816"

def find_repo_root(start=None):
    p = Path(start or Path.cwd()).resolve()
    for c in [p] + list(p.parents):
        if (c / "saju_engine.py").exists():
            return c
    raise FileNotFoundError("Run from inside the Chartpalja saju repo.")

def sha256_file(path):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for chunk in iter(lambda: f.read(1024*1024), b""):
            h.update(chunk)
    return h.hexdigest()

def bool_series(s):
    if s.dtype == bool:
        return s
    return s.astype(str).str.lower().map({"true":True, "false":False}).fillna(False)

ROOT = find_repo_root()
W2_ROSTER_DIR = ROOT / "research/ml/artifacts/v4_unified_dev_wave2_roster"
W2_EVENT_DIR = ROOT / "research/ml_corpus/v4_unified_dev_wave2_events"
W1_QA_DIR = ROOT / "research/ml/artifacts/v4_unified_dev_wave1"
W1_EVENT_DIR = ROOT / "research/ml_corpus/v4_unified_dev_events"
OUT = ROOT / "research/ml/artifacts/v4_unified_dev_wave2"
OUT.mkdir(parents=True, exist_ok=True)

ROSTER = W2_ROSTER_DIR / "V4_UNIFIED_DEV_WAVE2_SUBJECT_ROSTER_44.csv"
ROSTER_FREEZE = W2_ROSTER_DIR / "V4_UNIFIED_DEV_WAVE2_SUBJECT_ROSTER_FREEZE.json"
EVENTS = W2_EVENT_DIR / "V4_UNIFIED_DEV_WAVE2_VERIFIED_EVENTS.csv"
PAIRS_INPUT = W2_EVENT_DIR / "V4_UNIFIED_DEV_WAVE2_SELECTED_PAIRS.csv"
UNPAIRABLE = W2_EVENT_DIR / "V4_UNIFIED_DEV_WAVE2_UNPAIRABLE_LEDGER.csv"
COLLECTION_SUMMARY = W2_EVENT_DIR / "V4_UNIFIED_DEV_WAVE2_COLLECTION_SUMMARY.json"

# Prefer the 08 frozen Wave 1 output; fall back to the source-selected pairs only if needed.
W1_PAIRS = W1_QA_DIR / "V4_UNIFIED_DEV_WAVE1_FROZEN_PAIRS.csv"
if not W1_PAIRS.exists():
    W1_PAIRS = W1_EVENT_DIR / "V4_UNIFIED_DEV_WAVE1_SELECTED_PAIRS.csv"

for p in [ROSTER, ROSTER_FREEZE, EVENTS, PAIRS_INPUT, UNPAIRABLE, COLLECTION_SUMMARY, W1_PAIRS]:
    if not p.exists():
        raise FileNotFoundError(p)

roster = pd.read_csv(ROSTER)
events = pd.read_csv(EVENTS)
pairs_in = pd.read_csv(PAIRS_INPUT)
unpairable = pd.read_csv(UNPAIRABLE)
w1 = pd.read_csv(W1_PAIRS)
with open(ROSTER_FREEZE, encoding="utf-8") as f: roster_freeze = json.load(f)
with open(COLLECTION_SUMMARY, encoding="utf-8") as f: collection_summary = json.load(f)

assert roster_freeze["status"] == "V4_UNIFIED_DEV_WAVE2_SUBJECT_UNIVERSE_FROZEN"
assert len(roster) == 44 and roster.subject_id.nunique() == 44
assert roster_freeze["roster_sha256"] == sha256_file(ROSTER)
assert bool_series(roster["astrology_scored"]).sum() == 0
assert roster_freeze["astrology_scored"] is False
assert roster_freeze["event_collection_started"] is False
assert roster_freeze["membership_conditioned_on_chronology"] is False
assert roster_freeze["membership_conditioned_on_event_pairability"] is False
assert roster_freeze["membership_conditioned_on_astrology"] is False

print("repo:", ROOT)
print("Wave2 frozen roster:", len(roster))
print("Wave1 frozen pairs:", len(w1))
```

## 2. Source/taxonomy and frozen-membership integrity

```python

required_event = {
    "subject_id","name","preassigned_axis","event_year","polarity","event_type","description",
    "source_1","source_2","taxonomy_confirmed","polarity_confirmed","source_confirmed",
    "annual_eligible","collection_scope"
}
assert required_event.issubset(events.columns)
assert set(events.polarity) == {"positive","negative"}
assert events.event_id.nunique() == len(events)
assert events.subject_id.isin(roster.subject_id).all()
assert set(events.preassigned_axis).issubset({"COMPETITIVE","PROJECT","STATUS"})

for c in ["taxonomy_confirmed","polarity_confirmed","source_confirmed","annual_eligible"]:
    assert bool_series(events[c]).all(), f"{c} contains false values"
assert (events.collection_scope == "BOUNDED_MAJOR_EVENT_SOURCE_SWEEP").all()

assert (unpairable.status == "UNPAIRABLE_THIS_WAVE").all()
assert unpairable.subject_id.isin(roster.subject_id).all()
assert bool_series(unpairable["astrology_scored"]).sum() == 0
assert bool_series(unpairable["subject_membership_changed"]).sum() == 0

pairable_ids = set(events.subject_id)
unpairable_ids = set(unpairable.subject_id)
assert pairable_ids.isdisjoint(unpairable_ids)
assert pairable_ids | unpairable_ids == set(roster.subject_id), "Every one of the 44 frozen subjects must be accounted for."
assert len(pairable_ids) + len(unpairable_ids) == 44

# Each retained pairable subject must have both polarities and at least one distinct-year candidate.
g = events.groupby("subject_id")
for sid, x in g:
    assert {"positive","negative"}.issubset(set(x.polarity))
    pos = set(x.loc[x.polarity=="positive","event_year"].astype(int))
    neg = set(x.loc[x.polarity=="negative","event_year"].astype(int))
    assert any(a != b for a in pos for b in neg)

print("pairable:", len(pairable_ids), "unpairable:", len(unpairable_ids))
```

## 3. Deterministically reconstruct Wave 2 pairs from verified events

```python

def reconstruct_one(x):
    pos = x[x.polarity=="positive"].copy()
    neg = x[x.polarity=="negative"].copy()
    candidates = []
    for _, p in pos.iterrows():
        for _, n in neg.iterrows():
            py, ny = int(p.event_year), int(n.event_year)
            if py == ny:
                continue
            candidates.append((abs(py-ny), py, ny, p, n))
    if not candidates:
        raise RuntimeError(f"No distinct-year pair for {x.subject_id.iloc[0]}")
    candidates.sort(key=lambda z: (z[0], min(z[1],z[2]), max(z[1],z[2]), z[1], z[2]))
    gap, py, ny, p, n = candidates[0]
    return {
        "name": p["name"],
        "subject_id": p["subject_id"],
        "axis": p["preassigned_axis"],
        "positive_year": py,
        "positive_type": p["event_type"],
        "positive_description": p["description"],
        "negative_year": ny,
        "negative_type": n["event_type"],
        "negative_description": n["description"],
        "source_1": p["source_1"],
        "source_2": p["source_2"],
        "notes": p.get("notes",""),
        "pair_id": p["subject_id"] + "_P01",
        "year_gap": gap,
        "positive_earlier": py < ny,
        "source_1_domain": p["source_1_domain"],
        "source_2_domain": p["source_2_domain"],
        "independent_domain_sources": bool(p["independent_domain_sources"]),
        "taxonomy_confirmed": True,
        "polarity_confirmed": True,
        "source_confirmed": True,
        "annual_pair_eligible": True,
        "collection_scope": "BOUNDED_MAJOR_EVENT_SOURCE_SWEEP",
        "pair_selection_rule": "MIN_ABS_YEAR_GAP_WITHIN_VERIFIED_COLLECTED_CANDIDATES",
        "astrology_scored": False,
    }

recon = pd.DataFrame([reconstruct_one(x) for _, x in events.groupby("subject_id", sort=True)])
recon = recon.sort_values(["axis","subject_id"]).reset_index(drop=True)

# Compare only semantic pair columns; formatting/domain booleans are checked separately.
check_cols = [
    "subject_id","axis","positive_year","positive_type","negative_year","negative_type","year_gap"
]
expected = pairs_in[check_cols].copy().sort_values(["axis","subject_id"]).reset_index(drop=True)
actual = recon[check_cols].copy().sort_values(["axis","subject_id"]).reset_index(drop=True)
pd.testing.assert_frame_equal(actual, expected, check_dtype=False)

assert (recon.positive_year.astype(int) != recon.negative_year.astype(int)).all()
assert bool_series(recon["astrology_scored"]).sum() == 0

wave2_path = OUT / "V4_UNIFIED_DEV_WAVE2_FROZEN_PAIRS.csv"
recon.to_csv(wave2_path, index=False)

display(recon.groupby("axis").size().rename("wave2_pairs").to_frame())
```

## 4. Combine Wave 1 + Wave 2 and test predeclared axis targets

```python

TARGET = {"COMPETITIVE":30, "PROJECT":25, "STATUS":45}

# Harmonize columns and mark source wave.
w1 = w1.copy()
w2 = recon.copy()
w1["source_wave"] = 1
w2["source_wave"] = 2

common_cols = [c for c in w1.columns if c in w2.columns and c != "source_wave"]
combined = pd.concat(
    [w1[common_cols + ["source_wave"]], w2[common_cols + ["source_wave"]]],
    ignore_index=True
)

assert combined.pair_id.nunique() == len(combined)
assert bool_series(combined["astrology_scored"]).sum() == 0
assert (combined.positive_year.astype(int) != combined.negative_year.astype(int)).all()

counts = combined.groupby("axis").size().to_dict()
deficits = {a:max(0, TARGET[a]-int(counts.get(a,0))) for a in TARGET}
targets_met = all(v == 0 for v in deficits.values())

axis_audit = pd.DataFrame([
    {
        "axis": a,
        "target_pairs": TARGET[a],
        "wave1_pairs": int((w1.axis==a).sum()),
        "wave2_pairs": int((w2.axis==a).sum()),
        "combined_pairs": int(counts.get(a,0)),
        "deficit": deficits[a],
        "target_met": deficits[a] == 0,
    } for a in TARGET
])

combined_path = OUT / "V4_UNIFIED_DEV_COMBINED_FROZEN_PAIRS.csv"
combined.to_csv(combined_path, index=False)
axis_audit.to_csv(OUT / "V4_UNIFIED_DEV_COMBINED_AXIS_AUDIT.csv", index=False)

wave2_axis = pd.DataFrame([
    {
        "axis": a,
        "frozen_subjects": int((roster.preassigned_axis==a).sum()),
        "pairable": int((recon.axis==a).sum()),
        "unpairable": int((unpairable.preassigned_axis==a).sum()),
        "pairability": float((recon.axis==a).sum() / (roster.preassigned_axis==a).sum()),
    } for a in TARGET
])
wave2_axis.to_csv(OUT / "V4_UNIFIED_DEV_WAVE2_AXIS_AUDIT.csv", index=False)

display(axis_audit)
print("combined n:", len(combined), "targets_met:", targets_met)
```

## 5. Freeze decision — feature generation still requires a separate gate

```python

status = (
    "V4_UNIFIED_DEV_WAVE2_FROZEN_COMBINED_TARGETS_MET_READY_FOR_FEATURE_GENERATION_GATE"
    if targets_met else
    "V4_UNIFIED_DEV_WAVE2_FROZEN_NEEDS_ADDITIONAL_WAVE"
)

decision = {
    "version": "V4_UNIFIED_DEV_WAVE2_FREEZE_DECISION_V1",
    "notebook_version": NOTEBOOK_VERSION,
    "created_at": datetime.now().isoformat(timespec="seconds"),
    "status": status,
    "wave2": {
        "n_frozen_subjects": int(len(roster)),
        "n_pairs": int(len(recon)),
        "n_unpairable": int(len(unpairable)),
        "axis_counts": {a:int((recon.axis==a).sum()) for a in TARGET},
        "positive_earlier_share": float(recon.positive_earlier.mean()),
        "same_year_pairs": int((recon.positive_year.astype(int)==recon.negative_year.astype(int)).sum()),
        "astrology_scored": False,
    },
    "combined": {
        "n_pairs": int(len(combined)),
        "target_pairs": TARGET,
        "axis_counts": {a:int(counts.get(a,0)) for a in TARGET},
        "pair_deficits": deficits,
        "all_axis_targets_met": bool(targets_met),
        "astrology_scored": False,
    },
    "holdout_integrity": {
        "NEW_CONFIRM_loaded": False,
        "Validation_B_loaded": False,
        "Public_CHECK_loaded": False,
        "Public_FINAL_loaded": False,
    },
    "rules": {
        "wave2_subject_replacement": False,
        "directional_chronology_backfill": False,
        "astrology_feature_generation_performed": False,
        "astrology_feature_generation_allowed_in_this_notebook": False,
        "next_step_if_targets_met": "Create and freeze a separate combined-corpus feature-generation/preregistration gate before generating any astrology features.",
    },
}

with open(OUT / "V4_UNIFIED_DEV_WAVE2_FREEZE_DECISION.json","w",encoding="utf-8") as f:
    json.dump(decision,f,ensure_ascii=False,indent=2)

gate = {
    "version": "V4_UNIFIED_DEV_COMBINED_FEATURE_GATE_PRECHECK_V1",
    "status": "READY_TO_CREATE_FEATURE_GENERATION_GATE" if targets_met else "NOT_READY_MORE_EVENT_DATA_REQUIRED",
    "combined_pairs": int(len(combined)),
    "axis_counts": {a:int(counts.get(a,0)) for a in TARGET},
    "targets": TARGET,
    "all_targets_met": bool(targets_met),
    "astrology_feature_generation_allowed_now": False,
    "reason": "The event corpus may proceed to a separate preregistered feature-generation gate only after this freeze. This notebook intentionally does not expose astrology features.",
}
with open(OUT / "V4_UNIFIED_DEV_COMBINED_FEATURE_GATE_PRECHECK.json","w",encoding="utf-8") as f:
    json.dump(gate,f,ensure_ascii=False,indent=2)

print(json.dumps(decision, indent=2, ensure_ascii=False))
assert status.startswith("V4_UNIFIED_DEV_WAVE2_FROZEN")
```
