# 18A — V5 Identity Linkage Audit

**STOP EVENT RESEARCH BEFORE RUNNING THIS NOTEBOOK.**

Why this audit exists:

The V5 Wikidata expansion used a normalized-name join between a Wikidata role identity and the frozen AA birth snapshot. A same-name collision was later discovered in Batch 4 (`Walter Schmidt`).

This notebook audits identity linkage only.

It **does not**:
- load Batch 1-3 event outcomes,
- inspect pairability or chronology,
- generate astrology,
- inspect Control,
- replace any subject.

CONFIRM is audited locally for integrity, but detailed CONFIRM identity rows remain sealed and should not be sent back.


````python

from pathlib import Path
from datetime import datetime
import hashlib, json, re, time, unicodedata
import urllib.parse, urllib.request, urllib.error
import pandas as pd
import numpy as np
from IPython.display import display

NOTEBOOK_VERSION = "SAJU_ML_V5_IDENTITY_LINKAGE_AUDIT_20260817"

def find_repo_root(start=None):
    p=Path(start or Path.cwd()).resolve()
    for c in [p]+list(p.parents):
        if (c/"saju_engine.py").exists():
            return c
    raise FileNotFoundError("Run inside Chartpalja repo.")

def sha256_file(path):
    h=hashlib.sha256()
    with open(path,"rb") as f:
        for chunk in iter(lambda:f.read(1024*1024),b""):
            h.update(chunk)
    return h.hexdigest()

def norm_text(x):
    s=unicodedata.normalize("NFKD",str(x))
    s="".join(c for c in s if not unicodedata.combining(c))
    s=s.casefold()
    return re.sub(r"[^a-z0-9]+","",s)

ROOT=find_repo_root()
FINAL=ROOT/"research/ml/artifacts/v5_final_roster"
CORPUS=ROOT/"research/ml_corpus/v5_ground_truth"
OUT=ROOT/"research/ml/artifacts/v5_identity_linkage_audit"
CACHE=OUT/"wikidata_entity_cache"
OUT.mkdir(parents=True,exist_ok=True)
CACHE.mkdir(parents=True,exist_ok=True)

DEV_PATH=FINAL/"V5_DEV_SUBJECT_ROSTER_160.csv"
CONFIRM_PATH=FINAL/"V5_CONFIRM_SUBJECT_ROSTER_80_SEALED.csv"
POOL_PATH=FINAL/"V5_COMBINED_ROLE_ONLY_STRICT_AA_POOL.csv"
FREEZE_PATH=FINAL/"V5_FINAL_ROSTER_FREEZE_DECISION.json"
PROTOCOL_PATH=CORPUS/"V5_IDENTITY_LINKAGE_AUDIT_PROTOCOL.json"

for p in [DEV_PATH,CONFIRM_PATH,POOL_PATH,FREEZE_PATH,PROTOCOL_PATH]:
    if not p.exists(): raise FileNotFoundError(p)

dev=pd.read_csv(DEV_PATH)
confirm=pd.read_csv(CONFIRM_PATH)
pool=pd.read_csv(POOL_PATH)
freeze=json.load(open(FREEZE_PATH,encoding="utf-8"))
protocol=json.load(open(PROTOCOL_PATH,encoding="utf-8"))

assert freeze["status"]=="V5_FINAL_DEV_AND_CONFIRM_ROSTERS_FROZEN_READY_FOR_DEV_EVENT_COLLECTION"
assert sha256_file(DEV_PATH)==freeze["dev_roster_sha256"]
assert sha256_file(CONFIRM_PATH)==freeze["confirm_roster_sha256"]
assert protocol["status"]=="PREDECLARED_AFTER_OBJECTIVE_IDENTITY_COLLISION_DISCOVERY_BEFORE_REPAIR"
assert len(dev)==160 and len(confirm)==80
assert set(dev.name).isdisjoint(set(confirm.name))

print("IDENTITY AUDIT PREFLIGHT PASS")
print("DEV:",len(dev),"CONFIRM:",len(confirm),"candidate pool:",len(pool))
print("No event result path is defined or loaded.")
````

## 1. Fetch fixed Wikidata QIDs in batches

````python

API="https://www.wikidata.org/w/api.php"
BATCH=40

all_frames=[]
for scope_name,frame in [("DEV",dev),("CONFIRM",confirm),("POOL",pool)]:
    x=frame.copy()
    if "candidate_source" not in x:
        x["candidate_source"]=""
    if "wikidata_id" not in x:
        x["wikidata_id"]=np.nan
    q=x[
        (x.candidate_source=="V5_WIKIDATA_EXPANSION")
        & x.wikidata_id.notna()
    ].copy()
    all_frames.append(q)

qids=sorted(set(
    str(q).strip()
    for x in all_frames
    for q in x.wikidata_id
    if str(q).startswith("Q")
))
print("Unique Wikidata QIDs to audit:",len(qids))

def fetch_entities(ids, cache_path, attempts=4):
    if cache_path.exists():
        return json.load(open(cache_path,encoding="utf-8"))

    params={
        "action":"wbgetentities",
        "ids":"|".join(ids),
        "props":"claims|labels|descriptions",
        "languages":"en",
        "format":"json",
        "formatversion":"2",
        "origin":"*",
    }
    url=API+"?"+urllib.parse.urlencode(params)
    headers={
        "Accept":"application/json",
        "User-Agent":"Chartpalja-Saju-Research/1.0 identity-linkage-audit"
    }
    last=None
    for k in range(attempts):
        try:
            req=urllib.request.Request(url,headers=headers)
            with urllib.request.urlopen(req,timeout=90) as resp:
                obj=json.loads(resp.read().decode("utf-8"))
            json.dump(obj,open(cache_path,"w",encoding="utf-8"),ensure_ascii=False)
            return obj
        except Exception as e:
            last=e
            if k+1<attempts:
                time.sleep(4*(2**k))
    raise RuntimeError("Wikidata entity fetch failed: %r"%last)

entity_map={}
for i in range(0,len(qids),BATCH):
    ids=qids[i:i+BATCH]
    cp=CACHE/("entities_%04d_%04d.json"%(i,i+len(ids)-1))
    obj=fetch_entities(ids,cp)
    entity_map.update(obj.get("entities",{}))
    print("fetched",i,"-",i+len(ids)-1)

# Collect P19 place QIDs for label lookup.
place_qids=set()
for qid,e in entity_map.items():
    for cl in (e.get("claims",{}).get("P19") or []):
        try:
            v=cl["mainsnak"]["datavalue"]["value"]["id"]
            place_qids.add(v)
        except Exception:
            pass

place_qids=sorted(place_qids)
place_labels={}
for i in range(0,len(place_qids),BATCH):
    ids=place_qids[i:i+BATCH]
    cp=CACHE/("places_%04d_%04d.json"%(i,i+len(ids)-1))
    obj=fetch_entities(ids,cp)
    for qid,e in obj.get("entities",{}).items():
        place_labels[qid]=(e.get("labels",{}).get("en") or {}).get("value","")
print("birth-place labels:",len(place_labels))
````

## 2. Extract identity signals and classify linkage

````python

MALE_QID="Q6581097"
FEMALE_QID="Q6581072"

def parse_wd_time(v):
    # +1937-08-02T00:00:00Z
    t=str(v.get("time",""))
    m=re.match(r"^[+-](\d{4,})-(\d{2})-(\d{2})T",t)
    if not m: return None
    return "%04d-%02d-%02d"%tuple(map(int,m.groups()))

def entity_signals(qid):
    e=entity_map.get(str(qid),{})
    claims=e.get("claims",{})
    dobs=[]
    for cl in claims.get("P569",[]) or []:
        try:
            dv=cl["mainsnak"]["datavalue"]["value"]
            dobs.append({
                "date":parse_wd_time(dv),
                "precision":int(dv.get("precision",0)),
                "rank":cl.get("rank","normal")
            })
        except Exception:
            pass

    genders=[]
    for cl in claims.get("P21",[]) or []:
        try: genders.append(cl["mainsnak"]["datavalue"]["value"]["id"])
        except Exception: pass

    pids=[]
    for cl in claims.get("P19",[]) or []:
        try: pids.append(cl["mainsnak"]["datavalue"]["value"]["id"])
        except Exception: pass

    return {
        "wd_label":(e.get("labels",{}).get("en") or {}).get("value",""),
        "wd_description":(e.get("descriptions",{}).get("en") or {}).get("value",""),
        "wd_dobs":dobs,
        "wd_gender_qids":genders,
        "wd_birth_places":[place_labels.get(x,"") for x in pids],
    }

def local_city(place):
    return str(place).split(",")[0].strip()

def place_match(local_place, wd_places):
    a=norm_text(local_city(local_place))
    if not a: return None
    vals=[norm_text(x) for x in wd_places if str(x).strip()]
    if not vals: return None
    return any(a==b or (len(a)>=6 and (a in b or b in a)) for b in vals)

def gender_match(local_gender, qids):
    if not qids: return None
    g=str(local_gender).strip().lower()
    expected=MALE_QID if g.startswith("male") else FEMALE_QID if g.startswith("female") else None
    if expected is None: return None
    return expected in set(qids)

def classify(row):
    if str(row.get("candidate_source",""))!="V5_WIKIDATA_EXPANSION" or not str(row.get("wikidata_id","")).startswith("Q"):
        return "PASS_LEGACY_WAVE_A"

    s=entity_signals(row.wikidata_id)
    precise=[x["date"] for x in s["wd_dobs"] if x["date"] and x["precision"]>=11]
    exact=str(row.birth_date) in precise
    pm=place_match(row.birth_place,s["wd_birth_places"])
    gm=gender_match(row.gender,s["wd_gender_qids"])

    if gm is False:
        return "LIKELY_IDENTITY_MISMATCH"
    if exact:
        return "PASS_EXACT_DOB"
    if precise:
        if pm is True:
            return "REVIEW_DOB_DISCREPANCY_SAME_PLACE"
        if pm is False:
            return "LIKELY_IDENTITY_MISMATCH"
        return "REVIEW_DOB_DISCREPANCY_UNKNOWN_PLACE"
    return "REVIEW_MISSING_OR_LOW_PRECISION_DOB"

def audit_frame(frame,scope):
    out=frame.copy()
    sig=[]
    for _,r in out.iterrows():
        if str(r.get("wikidata_id","")).startswith("Q"):
            s=entity_signals(r.wikidata_id)
        else:
            s={"wd_label":"","wd_description":"","wd_dobs":[],"wd_gender_qids":[],"wd_birth_places":[]}
        sig.append(s)
    out["audit_scope"]=scope
    out["wd_label"]=[x["wd_label"] for x in sig]
    out["wd_description"]=[x["wd_description"] for x in sig]
    out["wd_dob_values"]=[
        "|".join(sorted(set(x["date"] for x in s["wd_dobs"] if x["date"])))
        for s in sig
    ]
    out["wd_dob_precisions"]=[
        "|".join(map(str,sorted(set(x["precision"] for x in s["wd_dobs"]))))
        for s in sig
    ]
    out["wd_birth_places"]=["|".join(s["wd_birth_places"]) for s in sig]
    out["wd_gender_qids"]=["|".join(s["wd_gender_qids"]) for s in sig]
    out["birth_place_match"]=[
        place_match(r.birth_place,s["wd_birth_places"])
        for (_,r),s in zip(out.iterrows(),sig)
    ]
    out["gender_match"]=[
        gender_match(r.gender,s["wd_gender_qids"])
        for (_,r),s in zip(out.iterrows(),sig)
    ]
    out["linkage_status"]=out.apply(classify,axis=1)
    return out

adev=audit_frame(dev,"DEV")
aconfirm=audit_frame(confirm,"CONFIRM")
apool=audit_frame(pool,"POOL")

display(adev.linkage_status.value_counts(dropna=False).rename("DEV_n"))
display(aconfirm.linkage_status.value_counts(dropna=False).rename("CONFIRM_n"))
````

## 3. Freeze audit outputs without repairing membership

````python

adev_path=OUT/"V5_IDENTITY_LINKAGE_AUDIT_DEV.csv"
aconfirm_path=OUT/"V5_IDENTITY_LINKAGE_AUDIT_CONFIRM_SEALED.csv"
apool_path=OUT/"V5_IDENTITY_LINKAGE_AUDIT_CANDIDATE_POOL.csv"

adev.to_csv(adev_path,index=False)
aconfirm.to_csv(aconfirm_path,index=False)
apool.to_csv(apool_path,index=False)

mismatch_dev=adev[adev.linkage_status=="LIKELY_IDENTITY_MISMATCH"].copy()
review_dev=adev[adev.linkage_status.str.startswith("REVIEW_")].copy()

mismatch_dev_path=OUT/"V5_IDENTITY_LINKAGE_MISMATCH_DEV.csv"
review_dev_path=OUT/"V5_IDENTITY_LINKAGE_REVIEW_DEV.csv"
mismatch_dev.to_csv(mismatch_dev_path,index=False)
review_dev.to_csv(review_dev_path,index=False)

selected=pd.concat([
    adev.assign(split_scope="DEV"),
    aconfirm.assign(split_scope="CONFIRM")
],ignore_index=True)

n_mismatch=int((selected.linkage_status=="LIKELY_IDENTITY_MISMATCH").sum())
n_review=int(selected.linkage_status.str.startswith("REVIEW_").sum())

if n_mismatch>0:
    status="V5_IDENTITY_LINKAGE_AUDIT_REPAIR_REQUIRED"
elif n_review>0:
    status="V5_IDENTITY_LINKAGE_AUDIT_MANUAL_REVIEW_REQUIRED"
else:
    status="V5_IDENTITY_LINKAGE_AUDIT_PASS"

pool_summary=(
    apool.groupby(["axis","linkage_status"])
    .size().rename("n").reset_index()
    if "axis" in apool.columns else
    apool.groupby(["linkage_status"]).size().rename("n").reset_index()
)

summary={
    "version":"V5_IDENTITY_LINKAGE_AUDIT_SUMMARY_V1",
    "created_at":datetime.now().isoformat(timespec="seconds"),
    "status":status,
    "trigger_subject":"V5DEV_022 Walter Schmidt",
    "DEV_counts":adev.linkage_status.value_counts().to_dict(),
    "CONFIRM_counts_aggregate_only":aconfirm.linkage_status.value_counts().to_dict(),
    "candidate_pool_counts_by_axis_and_status":pool_summary.to_dict(orient="records"),
    "DEV_likely_mismatch_n":int((adev.linkage_status=="LIKELY_IDENTITY_MISMATCH").sum()),
    "DEV_review_n":int(adev.linkage_status.str.startswith("REVIEW_").sum()),
    "CONFIRM_likely_mismatch_n":int((aconfirm.linkage_status=="LIKELY_IDENTITY_MISMATCH").sum()),
    "CONFIRM_review_n":int(aconfirm.linkage_status.str.startswith("REVIEW_").sum()),
    "rules":{
        "events_loaded":False,
        "pairability_loaded":False,
        "chronology_loaded":False,
        "astrology_loaded":False,
        "control_loaded":False,
        "membership_changed":False,
        "confirm_details_must_remain_sealed":True
    },
    "next_rule":(
        "If REPAIR_REQUIRED: stop event research and freeze a separate deterministic identity-repair protocol. "
        "If MANUAL_REVIEW_REQUIRED: resolve DEV review rows and aggregate CONFIRM review locally before any repair. "
        "Do not resume Batch 4 yet."
    )
}

summary_path=OUT/"V5_IDENTITY_LINKAGE_AUDIT_SUMMARY.json"
json.dump(summary,open(summary_path,"w",encoding="utf-8"),ensure_ascii=False,indent=2)

decision={
    "version":"V5_IDENTITY_LINKAGE_AUDIT_DECISION_V1",
    "status":status,
    "membership_changed":False,
    "DEV_audit_sha256":sha256_file(adev_path),
    "CONFIRM_sealed_audit_sha256":sha256_file(aconfirm_path),
    "candidate_pool_audit_sha256":sha256_file(apool_path),
    "summary_sha256":sha256_file(summary_path),
    "event_research_may_resume": status=="V5_IDENTITY_LINKAGE_AUDIT_PASS",
    "batch_04_may_resume": status=="V5_IDENTITY_LINKAGE_AUDIT_PASS"
}
decision_path=OUT/"V5_IDENTITY_LINKAGE_AUDIT_DECISION.json"
json.dump(decision,open(decision_path,"w",encoding="utf-8"),ensure_ascii=False,indent=2)

print(json.dumps(summary,ensure_ascii=False,indent=2))
print("\nDECISION:",status)
print("Do not send V5_IDENTITY_LINKAGE_AUDIT_CONFIRM_SEALED.csv.")
````
