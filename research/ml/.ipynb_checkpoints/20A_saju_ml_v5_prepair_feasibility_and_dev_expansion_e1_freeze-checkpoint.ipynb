# 20A — V5 Pre-pair Feasibility Gate + DEV Expansion E1 Freeze

This notebook deliberately runs **before any 1–5 year pair-gap inspection**.

The frozen pair gate requires at least:

- total pairable subjects: 80
- COMPETITIVE: 20
- PROJECT: 20
- STATUS: 25

A subject cannot possibly become pairable unless the frozen event corpus already contains
at least one positive and one negative eligible event for that subject. Therefore the
both-polarity subject count is a strict upper bound on later local pairability.

If that necessary condition already fails, this notebook does **not** generate pairs.
Instead, it freezes one new outcome-blind DEV expansion wave E1 using exact-DOB candidates only.

No astrology. No Control. No CONFIRM research.


````python

from pathlib import Path
from datetime import datetime
import hashlib, json, re, unicodedata
import numpy as np
import pandas as pd
from IPython.display import display

NOTEBOOK_VERSION="SAJU_ML_V5_PREPAIR_FEASIBILITY_EXPANSION_E1_20260817"
SEED=2026081703

def repo_root(start=None):
    p=Path(start or Path.cwd()).resolve()
    for c in [p]+list(p.parents):
        if (c/"saju_engine.py").exists():
            return c
    raise FileNotFoundError("Run inside Chartpalja repo.")

def sha256_file(path):
    h=hashlib.sha256()
    with open(path,"rb") as f:
        for chunk in iter(lambda:f.read(1024*1024),b""):
            h.update(chunk)
    return h.hexdigest()

def norm_name(x):
    s=unicodedata.normalize("NFKD",str(x))
    s="".join(c for c in s if not unicodedata.combining(c))
    s=s.casefold()
    return re.sub(r"[^a-z0-9]+","",s)

def selection_key(axis,qid,name):
    raw=f"{SEED}|{axis}|{qid}|{norm_name(name)}"
    return hashlib.sha256(raw.encode("utf-8")).hexdigest()

ROOT=repo_root()

FREEZE=ROOT/"research/ml/artifacts/v5_event_freeze"
REPAIR=ROOT/"research/ml/artifacts/v5_identity_repair"
AUDIT=ROOT/"research/ml/artifacts/v5_identity_linkage_audit"
CORPUS=ROOT/"research/ml_corpus/v5_ground_truth"

OUT=ROOT/"research/ml/artifacts/v5_dev_expansion_e1"
BATCH_OUT=OUT/"batches"
OUT.mkdir(parents=True,exist_ok=True)
BATCH_OUT.mkdir(parents=True,exist_ok=True)

FREEZE_DEC=FREEZE/"V5_DEV_EVENT_FREEZE_DECISION.json"
FREEZE_MAN=FREEZE/"V5_DEV_EVENT_FREEZE_MANIFEST.json"
SUBCOV=FREEZE/"V5_DEV_EVENT_FREEZE_SUBJECT_COVERAGE.csv"
DEV=REPAIR/"V5_DEV_SUBJECT_ROSTER_160_REPAIRED.csv"
CONF=REPAIR/"V5_CONFIRM_SUBJECT_ROSTER_80_REPAIRED_SEALED.csv"
POOL=AUDIT/"V5_IDENTITY_LINKAGE_AUDIT_CANDIDATE_POOL.csv"
GATE_PROTO=CORPUS/"V5_PREPAIR_FEASIBILITY_GATE_PROTOCOL.json"
EXP_PROTO=CORPUS/"V5_DEV_EXPANSION_WAVE_E1_PROTOCOL.json"

for p in [FREEZE_DEC,FREEZE_MAN,SUBCOV,DEV,CONF,POOL,GATE_PROTO,EXP_PROTO]:
    if not p.exists(): raise FileNotFoundError(p)

fdec=json.load(open(FREEZE_DEC,encoding="utf-8"))
fman=json.load(open(FREEZE_MAN,encoding="utf-8"))
gproto=json.load(open(GATE_PROTO,encoding="utf-8"))
eproto=json.load(open(EXP_PROTO,encoding="utf-8"))

assert fdec["status"]=="V5_DEV_EVENT_CORPUS_FROZEN_READY_FOR_LOCAL_PAIRING_NUISANCE_GATE"
assert fman["status"]==fdec["status"]
assert sha256_file(SUBCOV)==fman["subject_coverage_sha256"]
assert gproto["status"]=="PREDECLARED_BEFORE_ANY_1_TO_5_YEAR_PAIR_GAP_INSPECTION"
assert eproto["status"]=="PREDECLARED_AFTER_NECESSARY_CONDITION_FAILURE_BEFORE_EXPANSION_MEMBERSHIP_SELECTION"

sub=pd.read_csv(SUBCOV)
dev=pd.read_csv(DEV)
conf=pd.read_csv(CONF)
pool=pd.read_csv(POOL)

assert len(dev)==160 and len(conf)==80

print("20A PREFLIGHT PASS")
print("No pair file or pair-gap calculation has been loaded.")
````

## 1. Necessary-condition gate only

````python

sub["has_positive"]=sub.positive_events>0
sub["has_negative"]=sub.negative_events>0
sub["both_polarities"]=sub.has_positive & sub.has_negative

axis_both=(
    sub.groupby("preassigned_axis").both_polarities.sum()
    .reindex(["COMPETITIVE","PROJECT","STATUS"],fill_value=0)
    .astype(int)
)
total_both=int(sub.both_polarities.sum())

threshold_total=int(gproto["frozen_thresholds"]["min_pairable_subjects_total"])
threshold_axis=gproto["frozen_thresholds"]["min_pairable_subjects_by_axis"]

feasibility=pd.DataFrame([
    {
        "scope":"TOTAL",
        "both_polarity_upper_bound":total_both,
        "frozen_min_pairable":threshold_total,
        "necessary_condition_pass":total_both>=threshold_total
    },
    *[
        {
            "scope":axis,
            "both_polarity_upper_bound":int(axis_both[axis]),
            "frozen_min_pairable":int(threshold_axis[axis]),
            "necessary_condition_pass":int(axis_both[axis])>=int(threshold_axis[axis])
        }
        for axis in ["COMPETITIVE","PROJECT","STATUS"]
    ]
])

display(feasibility)

necessary_pass=bool(feasibility.necessary_condition_pass.all())
if necessary_pass:
    status="V5_PREPAIR_FEASIBILITY_NECESSARY_CONDITION_PASS_PROCEED_TO_PAIR_GENERATION"
else:
    status="V5_PREPAIR_FEASIBILITY_FAIL_EXPANSION_REQUIRED"

feas_path=OUT/"V5_PREPAIR_FEASIBILITY_SUMMARY.csv"
feasibility.to_csv(feas_path,index=False)

gate_decision={
    "version":"V5_PREPAIR_FEASIBILITY_DECISION_V1",
    "created_at":datetime.now().isoformat(timespec="seconds"),
    "status":status,
    "both_polarity_upper_bound_total":total_both,
    "both_polarity_upper_bound_by_axis":axis_both.to_dict(),
    "frozen_min_pairable_total":threshold_total,
    "frozen_min_pairable_by_axis":threshold_axis,
    "pair_gap_inspected":False,
    "pairs_generated":False,
    "astrology_generated":False,
    "control_scored":False,
    "confirm_researched":False,
    "feasibility_summary_sha256":sha256_file(feas_path),
    "next_rule":(
        "Freeze and collect a whole new DEV expansion wave under V5_DEV_EXPANSION_WAVE_E1_PROTOCOL."
        if not necessary_pass else
        "Proceed to deterministic local 1-5y pair generation."
    )
}
gate_dec_path=OUT/"V5_PREPAIR_FEASIBILITY_DECISION.json"
json.dump(gate_decision,open(gate_dec_path,"w",encoding="utf-8"),ensure_ascii=False,indent=2)

print(json.dumps(gate_decision,ensure_ascii=False,indent=2))

# This notebook version is intended for the currently observed necessary-condition failure.
assert status=="V5_PREPAIR_FEASIBILITY_FAIL_EXPANSION_REQUIRED", (
    "Necessary condition unexpectedly passed. Stop here and build the pair-generation notebook instead."
)
````

## 2. Freeze one whole new exact-DOB DEV expansion wave E1

````python

# Candidate identities must already have exact day-level DOB linkage verification.
cand=pool[
    (pool.linkage_status=="PASS_EXACT_DOB")
    & pool.wikidata_id.notna()
].copy()

cand["norm_name_e1"]=cand.name.map(norm_name)

selected_names=set(dev.name.map(norm_name)) | set(conf.name.map(norm_name))
selected_qids=set(
    pd.concat([dev.wikidata_id,conf.wikidata_id])
    .dropna().astype(str)
)

cand=cand[
    ~cand.norm_name_e1.isin(selected_names)
    & ~cand.wikidata_id.astype(str).isin(selected_qids)
].copy()

cand["selection_key_e1"]=cand.apply(
    lambda r:selection_key(str(r.axis),str(r.wikidata_id),r["name"]),
    axis=1
)

cand=(
    cand.sort_values(["axis","selection_key_e1","wikidata_id","norm_name_e1"])
    .drop_duplicates("wikidata_id")
    .drop_duplicates(["axis","norm_name_e1"])
    .reset_index(drop=True)
)

quotas=eproto["wave"]["axis_quotas"]
availability=cand.axis.value_counts().to_dict()

suff=pd.DataFrame([
    {
        "axis":axis,
        "quota":int(quotas[axis]),
        "available_exact_DOB_candidates":int(availability.get(axis,0)),
        "sufficient":int(availability.get(axis,0))>=int(quotas[axis])
    }
    for axis in ["COMPETITIVE","PROJECT","STATUS"]
])
display(suff)
assert suff.sufficient.all()

chosen_parts=[]
for axis in ["COMPETITIVE","PROJECT","STATUS"]:
    x=(
        cand[cand.axis==axis]
        .sort_values(["selection_key_e1","wikidata_id","norm_name_e1"])
        .head(int(quotas[axis]))
        .copy()
    )
    assert len(x)==int(quotas[axis])
    chosen_parts.append(x)

chosen=pd.concat(chosen_parts,ignore_index=True)

assert len(chosen)==160
assert chosen.wikidata_id.nunique()==160
assert chosen.norm_name_e1.nunique()==160
assert set(chosen.wikidata_id.astype(str)).isdisjoint(selected_qids)
````

## 3. Assign fixed subject IDs and the same 8-batch axis schedule

````python

# Same batch composition as original 160-person DEV.
batch_specs={
    1:{"COMPETITIVE":5,"PROJECT":7,"STATUS":9},
    2:{"COMPETITIVE":5,"PROJECT":7,"STATUS":9},
    3:{"COMPETITIVE":5,"PROJECT":6,"STATUS":9},
    4:{"COMPETITIVE":5,"PROJECT":6,"STATUS":9},
    5:{"COMPETITIVE":5,"PROJECT":6,"STATUS":9},
    6:{"COMPETITIVE":5,"PROJECT":6,"STATUS":9},
    7:{"COMPETITIVE":5,"PROJECT":6,"STATUS":8},
    8:{"COMPETITIVE":5,"PROJECT":6,"STATUS":8},
}
assert sum(v["COMPETITIVE"] for v in batch_specs.values())==40
assert sum(v["PROJECT"] for v in batch_specs.values())==50
assert sum(v["STATUS"] for v in batch_specs.values())==70

queues={
    axis:chosen[chosen.axis==axis].sort_values(
        ["selection_key_e1","wikidata_id"]
    ).reset_index(drop=True)
    for axis in ["COMPETITIVE","PROJECT","STATUS"]
}
ptr={k:0 for k in queues}

schedule=[]
seq=1
for b in range(1,9):
    order=1
    # Preserve a deterministic mixed-axis order instead of grouping an entire batch by axis.
    remaining=batch_specs[b].copy()
    axis_cycle=["COMPETITIVE","PROJECT","STATUS"]
    while sum(remaining.values())>0:
        for axis in axis_cycle:
            if remaining[axis]<=0:
                continue
            r=queues[axis].iloc[ptr[axis]]
            ptr[axis]+=1
            remaining[axis]-=1

            rec=r.to_dict()
            rec.update({
                "subject_id":f"V5DEVX_{seq:03d}",
                "split":"DEV_EXPANSION_E1",
                "preassigned_axis":axis,
                "expansion_wave":"E1",
                "batch_id":b,
                "research_order_in_batch":order,
                "event_collection_started":False,
                "astrology_scored":False,
            })
            schedule.append(rec)
            seq+=1
            order+=1

roster=pd.DataFrame(schedule)

assert len(roster)==160
assert roster.subject_id.nunique()==160
assert roster.preassigned_axis.value_counts().to_dict()=={
    "STATUS":70,"PROJECT":50,"COMPETITIVE":40
}
assert roster.batch_id.value_counts().sort_index().to_dict()=={
    1:21,2:21,3:20,4:20,5:20,6:20,7:19,8:19
}

# Remove helper-only columns from published roster.
for c in ["norm_name_e1"]:
    if c in roster.columns:
        roster=roster.drop(columns=c)

roster_path=OUT/"V5_DEV_EXPANSION_E1_ROSTER_160.csv"
roster.to_csv(roster_path,index=False)

work_cols=[
    "batch_id","research_order_in_batch","subject_id","name",
    "preassigned_axis","candidate_source","birth_date","birth_place",
    "gender","wikidata_id","rodden_rating","source_row_key"
]
work_cols=[c for c in work_cols if c in roster.columns]
work=roster[work_cols].copy()

work_path=OUT/"V5_DEV_EXPANSION_E1_EVENT_WORKLIST_8BATCH.csv"
work.to_csv(work_path,index=False)

for b in range(1,9):
    wb=work[work.batch_id==b].sort_values("research_order_in_batch").copy()
    wb.to_csv(
        BATCH_OUT/f"V5_DEV_EXPANSION_E1_BATCH_{b:02d}_SUBJECTS.csv",
        index=False
    )

print("Expansion E1 roster frozen:",len(roster))
display(
    work.groupby(["batch_id","preassigned_axis"]).size().unstack(fill_value=0)
)
````

## 4. Freeze decision

````python

decision={
    "version":"V5_DEV_EXPANSION_E1_FREEZE_DECISION_V1",
    "notebook_version":NOTEBOOK_VERSION,
    "created_at":datetime.now().isoformat(timespec="seconds"),
    "status":"V5_DEV_EXPANSION_E1_ROSTER_FROZEN_READY_FOR_EVENT_COLLECTION",
    "trigger_gate_status":"V5_PREPAIR_FEASIBILITY_FAIL_EXPANSION_REQUIRED",
    "n_subjects":160,
    "axis_counts":{"COMPETITIVE":40,"PROJECT":50,"STATUS":70},
    "roster_sha256":sha256_file(roster_path),
    "worklist_sha256":sha256_file(work_path),
    "prepair_feasibility_decision_sha256":sha256_file(gate_dec_path),
    "prepair_gate_protocol_sha256":sha256_file(GATE_PROTO),
    "expansion_protocol_sha256":sha256_file(EXP_PROTO),
    "selection_seed":SEED,
    "rules":{
        "exact_DOB_candidates_only":True,
        "current_DEV_or_CONFIRM_identity_reused":False,
        "event_outcomes_used_for_membership":False,
        "pairability_used_for_membership":False,
        "pair_gap_used_for_membership":False,
        "chronology_used_for_membership":False,
        "astrology_used_for_membership":False,
        "control_used_for_membership":False,
        "confirm_remains_sealed":True
    },
    "event_collection_may_begin":True,
    "pair_generation_allowed":False,
    "astrology_generation_allowed":False,
    "next_rule":(
        "Collect E1 events under the original frozen bounded source-sweep axis contract. "
        "Do not inspect 1-5y pair gaps until E1 event collection is frozen."
    )
}
decision_path=OUT/"V5_DEV_EXPANSION_E1_FREEZE_DECISION.json"
json.dump(decision,open(decision_path,"w",encoding="utf-8"),ensure_ascii=False,indent=2)

print(json.dumps(decision,ensure_ascii=False,indent=2))
````
