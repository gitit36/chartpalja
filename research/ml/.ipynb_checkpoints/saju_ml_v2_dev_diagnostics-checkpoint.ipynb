# SAJU ML V2 External DEV Diagnostics

## Purpose

The predeclared V2 NEW_DEV gate produced:

- no surviving candidate
- TENGOD remained directionally promising
- SHINSAL_ALL weakened

This notebook is **diagnostic only**.

It must not:

- change the V2 gate
- tune coefficients
- add new astrology features
- open NEW_CONFIRM
- load Public CHECK / FINAL
- alter the winner freeze

It answers:

1. Is TENGOD significantly above chance on the external NEW_DEV set?
2. Is TENGOD consistently better than Production Control, or driven by a few subjects?
3. Does performance differ between truly fresh V2 subjects and prior-unused public candidates?
4. Does performance survive high-confidence-only event filtering?
5. Is there obvious feature-distribution drift between V1 training and V2 NEW_DEV?

````python
from pathlib import Path
from itertools import product
import json
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.base import clone
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

warnings.filterwarnings("ignore")

SEED = 20260815
N_PERMUTATIONS = 2000
N_BOOTSTRAP = 20000

def find_repo_root(start=None):
    p = Path(start or Path.cwd()).resolve()
    for candidate in [p, *p.parents]:
        if (candidate / "saju_engine.py").exists():
            return candidate
    raise FileNotFoundError("Run from inside the saju repo.")

REPO_ROOT = find_repo_root()

V1_CORPUS_PATH = REPO_ROOT / "research/ml_corpus/v1/SAJU_ML_CORPUS_V1.json"
V1_FEATURE_DIR = REPO_ROOT / "research/ml/artifacts/v1"
V2_DEV_PATH = REPO_ROOT / "research/ml_corpus/v2/SAJU_ML_CORPUS_V2_NEW_DEV.json"
V2_FEATURE_PATH = REPO_ROOT / "research/ml/artifacts/v2/V2_NEW_DEV_feature_table_R2.csv"
V2_RESULTS_PATH = REPO_ROOT / "research/ml/artifacts/v2/V2_DEV_candidate_results.csv"
V2_FREEZE_PATH = REPO_ROOT / "research/ml/artifacts/v2/V2_DEV_WINNER_FREEZE.json"
CANDIDATE_FREEZE_PATH = REPO_ROOT / "research/ml_corpus/v2/SAJU_ML_V2_CANDIDATE_FREEZE.json"

OUT = REPO_ROOT / "research/ml/artifacts/v2_diagnostics"
OUT.mkdir(parents=True, exist_ok=True)

print("repo:", REPO_ROOT)
print("output:", OUT)
````

## 1. Guardrails and frozen result

````python
freeze = json.load(open(V2_FREEZE_PATH, encoding="utf-8"))
assert freeze["status"] == "V2_DEV_NO_CANDIDATE_SURVIVES"
assert freeze["winner"] is None
assert freeze["integrity"]["confirm_file_loaded"] is False

dev_results = pd.read_csv(V2_RESULTS_PATH)
display(dev_results)

print("Frozen status:", freeze["status"])
print("NEW_CONFIRM loaded:", freeze["integrity"]["confirm_file_loaded"])
````

## 2. Rebuild V1 and V2 Career ranking rows

````python
def build_target_rows(corpus):
    rows = []
    for s in corpus["subjects"]:
        if not s["birth"].get("birth_time_known", True):
            continue
        for e in s["events"]:
            if not e.get("ml_usable", True):
                continue
            lab = e["targets"].get("Career", "unknown")
            if lab not in {"positive", "negative"}:
                continue
            rows.append({
                "subject_id": s["subject_id"],
                "name": s["name"],
                "origin_group": s.get("origin_group", s.get("source_pool")),
                "year": int(e["year"]),
                "label": 1 if lab == "positive" else 0,
                "confidence": e.get("source_event_confidence", "unknown"),
            })

    raw = pd.DataFrame(rows)
    collapsed = []

    for (sid, year), g in raw.groupby(["subject_id", "year"]):
        if g["label"].nunique() > 1:
            continue

        # use the highest available confidence for the year
        conf_order = {"high": 2, "medium": 1, "low": 0, "unknown": -1}
        best_conf = max(
            g["confidence"].astype(str),
            key=lambda x: conf_order.get(x, -1)
        )

        collapsed.append({
            "subject_id": sid,
            "name": g["name"].iloc[0],
            "origin_group": g["origin_group"].iloc[0],
            "year": int(year),
            "label": int(g["label"].iloc[0]),
            "confidence": best_conf,
        })

    out = pd.DataFrame(collapsed)

    pairable = []
    for sid, g in out.groupby("subject_id"):
        if set(g["label"]) == {0, 1}:
            pairable.append(sid)

    return out[out["subject_id"].isin(pairable)].copy()

v1_corpus = json.load(open(V1_CORPUS_PATH, encoding="utf-8"))
v2_corpus = json.load(open(V2_DEV_PATH, encoding="utf-8"))

v1_targets = build_target_rows(v1_corpus)
v2_targets = build_target_rows(v2_corpus)

v1_pq = V1_FEATURE_DIR / "ml_feature_table_v1.parquet"
v1_csv = V1_FEATURE_DIR / "ml_feature_table_v1.csv"

if v1_pq.exists():
    v1_feat = pd.read_parquet(v1_pq)
else:
    v1_feat = pd.read_csv(v1_csv)

v2_feat = pd.read_csv(V2_FEATURE_PATH)

v1_df = v1_targets.merge(
    v1_feat, on=["subject_id", "year"], how="inner", validate="one_to_one"
)
v2_df = v2_targets.merge(
    v2_feat, on=["subject_id", "year"], how="inner", validate="one_to_one"
)

print("V1 pairable subjects:", v1_df["subject_id"].nunique())
print("V2 pairable subjects:", v2_df["subject_id"].nunique())
print(v2_df["origin_group"].value_counts())
````

## 3. Frozen candidates only

````python
candidate_freeze = json.load(open(CANDIDATE_FREEZE_PATH, encoding="utf-8"))

TENGOD_COLS = candidate_freeze["candidates"]["TENGOD"]["features"]
SHINSAL_COLS = candidate_freeze["candidates"]["SHINSAL_ALL"]["features"]

MODEL = Pipeline([
    ("scale", StandardScaler()),
    ("model", LogisticRegression(
        penalty="l2",
        C=0.3,
        max_iter=3000,
        random_state=SEED,
    )),
])

def make_pairwise_training(df, cols):
    Xs, ys = [], []
    for sid, g in df.groupby("subject_id"):
        pos = g[g["label"] == 1]
        neg = g[g["label"] == 0]
        for _, p in pos.iterrows():
            for _, n in neg.iterrows():
                d = p[cols].to_numpy(float) - n[cols].to_numpy(float)
                Xs.append(d); ys.append(1)
                Xs.append(-d); ys.append(0)
    return np.asarray(Xs), np.asarray(ys)

def fit_model(df, cols):
    X, y = make_pairwise_training(df, cols)
    m = clone(MODEL)
    m.fit(X, y)
    return m

def score_model(m, df, cols):
    scaler = m.named_steps["scale"]
    clf = m.named_steps["model"]
    X = scaler.transform(df[cols].to_numpy(float))
    return X @ clf.coef_.ravel()

def subject_pairwise(df, score_col):
    rows = []
    for sid, g in df.groupby("subject_id"):
        pos = g.loc[g["label"] == 1, score_col].to_numpy(float)
        neg = g.loc[g["label"] == 0, score_col].to_numpy(float)
        if len(pos) == 0 or len(neg) == 0:
            continue
        vals = [
            1.0 if p > n else 0.5 if p == n else 0.0
            for p, n in product(pos, neg)
        ]
        rows.append({
            "subject_id": sid,
            "pairwise": float(np.mean(vals)),
            "hit": float(np.mean(pos) > np.mean(neg)),
            "sep": float(np.mean(pos) - np.mean(neg)),
        })
    return pd.DataFrame(rows)

models = {
    "TENGOD": fit_model(v1_df, TENGOD_COLS),
    "SHINSAL_ALL": fit_model(v1_df, SHINSAL_COLS),
}

scored = v2_df[
    ["subject_id","name","origin_group","year","label","confidence","baseline__control_score"]
].copy()

for name, cols in [("TENGOD", TENGOD_COLS), ("SHINSAL_ALL", SHINSAL_COLS)]:
    scored[name] = score_model(models[name], v2_df, cols)

scored.to_csv(OUT / "V2_DEV_scored_rows_diagnostic.csv", index=False)
````

## 4. Fixed-model permutation test vs chance

The model is frozen from V1.

We do **not retrain** during permutations.

Within each V2 subject, labels are shuffled while preserving that subject's event-year scores.
This tests whether the external ranking is stronger than random label assignment.

````python
def macro_pairwise(df, score_col):
    s = subject_pairwise(df, score_col)
    return float(s["pairwise"].mean())

def fixed_model_permutation(df, score_col, n_perm=N_PERMUTATIONS):
    observed = macro_pairwise(df, score_col)
    rng = np.random.default_rng(SEED + sum(map(ord, score_col)))
    null = []

    for _ in range(n_perm):
        p = df.copy()
        for sid, idx in p.groupby("subject_id").groups.items():
            idx = list(idx)
            labels = p.loc[idx, "label"].to_numpy().copy()
            rng.shuffle(labels)
            p.loc[idx, "label"] = labels
        null.append(macro_pairwise(p, score_col))

    null = np.asarray(null)
    pval = (1 + np.sum(null >= observed)) / (len(null) + 1)

    return {
        "candidate": score_col,
        "observed": observed,
        "null_mean": float(null.mean()),
        "null_p95": float(np.quantile(null, 0.95)),
        "p_value": float(pval),
    }

perm_rows = [
    fixed_model_permutation(scored, "TENGOD"),
    fixed_model_permutation(scored, "SHINSAL_ALL"),
]

perm_df = pd.DataFrame(perm_rows)
display(perm_df)
perm_df.to_csv(OUT / "V2_DEV_fixed_model_permutation.csv", index=False)
````

## 5. Subject bootstrap vs chance and vs Control

````python
def bootstrap_macro(subject_df, n=N_BOOTSTRAP):
    vals = subject_df["pairwise"].to_numpy(float)
    rng = np.random.default_rng(SEED + 77)
    boot = np.array([
        rng.choice(vals, size=len(vals), replace=True).mean()
        for _ in range(n)
    ])
    return {
        "mean": float(vals.mean()),
        "ci_low": float(np.quantile(boot, .025)),
        "ci_high": float(np.quantile(boot, .975)),
        "p_gt_half": float(np.mean(boot > .5)),
    }

control_subj = subject_pairwise(scored, "baseline__control_score")

summary_rows = []
paired_tables = {}

for cand in ("TENGOD", "SHINSAL_ALL"):
    s = subject_pairwise(scored, cand)
    chance = bootstrap_macro(s)

    paired = (
        s[["subject_id","pairwise"]]
        .rename(columns={"pairwise":"candidate"})
        .merge(
            control_subj[["subject_id","pairwise"]].rename(
                columns={"pairwise":"control"}
            ),
            on="subject_id"
        )
    )
    paired["delta"] = paired["candidate"] - paired["control"]

    vals = paired["delta"].to_numpy(float)
    rng = np.random.default_rng(SEED + 99 + len(cand))
    boot = np.array([
        rng.choice(vals, size=len(vals), replace=True).mean()
        for _ in range(N_BOOTSTRAP)
    ])

    summary_rows.append({
        "candidate": cand,
        "macro_pairwise": chance["mean"],
        "chance_ci_low": chance["ci_low"],
        "chance_ci_high": chance["ci_high"],
        "bootstrap_p_pairwise_gt_0_5": chance["p_gt_half"],
        "delta_vs_control": float(vals.mean()),
        "delta_ci_low": float(np.quantile(boot,.025)),
        "delta_ci_high": float(np.quantile(boot,.975)),
        "bootstrap_p_delta_gt_0": float(np.mean(boot > 0)),
        "subjects_better": int((vals > 0).sum()),
        "subjects_equal": int((vals == 0).sum()),
        "subjects_worse": int((vals < 0).sum()),
    })

    paired_tables[cand] = paired
    paired.to_csv(OUT / f"{cand}_subject_deltas_diagnostic.csv", index=False)

summary_df = pd.DataFrame(summary_rows)
display(summary_df)
summary_df.to_csv(OUT / "V2_DEV_bootstrap_diagnostic.csv", index=False)
````

## 6. Performance by origin group

````python
group_rows = []

for cand in ("TENGOD","SHINSAL_ALL","baseline__control_score"):
    for group, g in scored.groupby("origin_group"):
        s = subject_pairwise(g, cand)
        if s.empty:
            continue
        group_rows.append({
            "score": cand,
            "origin_group": group,
            "n_subjects": len(s),
            "macro_pairwise": float(s["pairwise"].mean()),
            "hit_rate": float(s["hit"].mean()),
        })

group_df = pd.DataFrame(group_rows)
display(group_df.sort_values(["origin_group","macro_pairwise"], ascending=[True,False]))
group_df.to_csv(OUT / "V2_DEV_origin_group_diagnostic.csv", index=False)
````

## 7. High-confidence-only sensitivity

A subject remains evaluable only if both positive and negative high-confidence years remain.

````python
high = scored[scored["confidence"] == "high"].copy()

high_rows = []
for cand in ("TENGOD","SHINSAL_ALL","baseline__control_score"):
    s = subject_pairwise(high, cand)
    if s.empty:
        continue
    high_rows.append({
        "score": cand,
        "n_subjects": len(s),
        "macro_pairwise": float(s["pairwise"].mean()),
        "hit_rate": float(s["hit"].mean()),
    })

high_df = pd.DataFrame(high_rows)
display(high_df.sort_values("macro_pairwise", ascending=False))
high_df.to_csv(OUT / "V2_DEV_high_conf_diagnostic.csv", index=False)
````

## 8. Subject-level concentration

````python
for cand in ("TENGOD","SHINSAL_ALL"):
    paired = paired_tables[cand]
    print("\n", cand)
    print("better/equal/worse:",
          int((paired["delta"]>0).sum()),
          int((paired["delta"]==0).sum()),
          int((paired["delta"]<0).sum()))
    display(paired.sort_values("delta").head(10))
    display(paired.sort_values("delta", ascending=False).head(10))
````

## 9. Ten-God feature drift V1 → V2

Large shifts in categorical prevalence can explain a model's attenuation even when its learned relation is real.

````python
drift_rows = []

for c in TENGOD_COLS:
    a = pd.to_numeric(v1_df[c], errors="coerce").fillna(0.0)
    b = pd.to_numeric(v2_df[c], errors="coerce").fillna(0.0)

    drift_rows.append({
        "feature": c,
        "v1_mean": float(a.mean()),
        "v2_mean": float(b.mean()),
        "abs_mean_shift": float(abs(a.mean()-b.mean())),
        "v1_nonzero": float((a!=0).mean()),
        "v2_nonzero": float((b!=0).mean()),
    })

drift_df = pd.DataFrame(drift_rows).sort_values(
    "abs_mean_shift", ascending=False
)

display(drift_df)
drift_df.to_csv(OUT / "V2_DEV_tengod_feature_drift.csv", index=False)
````

## 10. Diagnostic decision

This diagnostic **cannot overturn** `V2_DEV_NO_CANDIDATE_SURVIVES`.

Suggested interpretation:

### `COLLECT_MORE_FRESH_DEV_AROUND_TENGOD`
Use when:
- TENGOD external pairwise remains > 0.55
- fixed-model permutation is suggestive
- signal is not isolated to one source group
- SHINSAL is clearly weaker or less stable

### `RETHINK_BOTH_CANDIDATES`
Use when:
- TENGOD is near chance after diagnostic sensitivity checks

In either case:

> NEW_CONFIRM remains sealed.

````python
tg = dev_results.loc[dev_results["candidate"]=="TENGOD"].iloc[0]
tg_perm = perm_df.loc[perm_df["candidate"]=="TENGOD"].iloc[0]
sh = dev_results.loc[dev_results["candidate"]=="SHINSAL_ALL"].iloc[0]

if (
    tg["macro_pairwise"] > 0.55
    and tg_perm["p_value"] < 0.10
    and tg["macro_pairwise"] > sh["macro_pairwise"]
):
    diagnostic_status = "COLLECT_MORE_FRESH_DEV_AROUND_TENGOD"
else:
    diagnostic_status = "RETHINK_BOTH_CANDIDATES"

decision = {
    "frozen_v2_status": freeze["status"],
    "diagnostic_status": diagnostic_status,
    "confirmation_must_remain_sealed": True,
    "note": (
        "Diagnostic result does not modify the predeclared V2 gate. "
        "No retuning or confirmation scoring was performed."
    ),
}

print(json.dumps(decision, indent=2))
json.dump(
    decision,
    open(OUT / "V2_DEV_DIAGNOSTIC_DECISION.json","w",encoding="utf-8"),
    indent=2,
)
````
